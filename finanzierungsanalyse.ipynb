## 2. Externe Prognose, Datenquelle und Planungshorizont

### Warum eine externe Prognose?

Die Aufgabe in diesem Notebook ist, aus vorliegenden wirtschaftlichen
Informationen eine Finanzierungsentscheidung abzuleiten. Für den
Inflationspfad verwenden wir deshalb eine veröffentlichte Prognose als
**Modelleingabe**.

Das Statistische Bundesamt (Destatis) veröffentlicht beobachtete
Preisentwicklungen. Für den zukünftigen Inflationspfad unseres
Deutschland-Falls verwenden wir eine veröffentlichte Prognose der Deutschen
Bundesbank. Projektionen der EZB beziehen sich auf den Euroraum und können
später zur Einordnung dienen; sie werden nicht ohne Weiteres mit einer
Deutschland-Prognose gleichgesetzt.

### Quelle und zeitlicher Stand

| Festlegung | Wert |
|---|---|
| Beratungsstichtag dieses Beispiels | 24. September 2026 |
| Externe Prognose | Deutschland-Prognose der Deutschen Bundesbank |
| Veröffentlichungsdatum der verwendeten Prognose | 12. Juni 2026 |
| Preismaß | Harmonisierter Verbraucherpreisindex (HVPI), Deutschland |
| Planjahre | ahre mit Werten aus der Bundesbank-Prognose |
| Einheit der veröffentlichten Werte | Jährliche Inflationsrate in Prozent |

**Quelle:** [Deutsche Bundesbank, Deutschland-Prognose vom
12. Juni 2026](https://www.bundesbank.de/de/presse/pressemitteilungen/deutschland-prognose-der-bundesbank-energiepreis-schock-bremst-konjunkturerholung-964728).

Das Entscheidungsmodell betrachtet sechs vollständige Planjahre, von 2027
bis 2032. Für 2027 und 2028 übernehmen wir die veröffentlichten Werte der
Bundesbank. Für 2029 bis 2032 setzen wir ausdrücklich gekennzeichnete
Szenarioannahmen ein. Diese Annahmen sind keine amtlichen Prognosen.

### Bedeutung der Inflationswerte

Die veröffentlichten Jahresraten vergleichen den **Jahresdurchschnitt**
des Preisindex mit dem Durchschnitt des Vorjahres. Sie sind keine
Prognosen für den Preisstand an einem bestimmten Tag.

Aus der allgemeinen Verbraucherpreisinflation folgt weder unmittelbar,
wie stark die Material- oder Lohnkosten dieses Unternehmens steigen,
noch welcher Kreditzins gelten wird. Diese Beziehungen und die
Kreditkonditionen beschreiben wir später im Unternehmens- und
Finanzierungsmodell.

Die Bundesbank-Prognose enthält bereits Annahmen zu einem
Energiepreisschock. Ein späteres Krisenszenario muss daher als
**zusätzliche Abweichung gegenüber dieser Prognose** definiert werden.
So berücksichtigen wir denselben Schock nicht zweimal.

```python
import pandas as pd

prognose_info = {
    "quelle": "Deutsche Bundesbank",
    "veroeffentlicht_am": "2026-06-12",
    "preisindex": "HVPI Deutschland",
    "einheit": "jaehrliche Durchschnittsrate in Prozent",
}

inflationspfad = pd.DataFrame({
    "jahr": [2027, 2028],
    "inflation_prozent": [2.7, 1.9],
})

inflationspfad
```

## Kumulierte Preisentwicklung
### Preisindex aus der Inflationsprognose

Wir setzen den durchschnittlichen Preisindex für 2026 auf 100. Für die
folgenden Jahre gilt:

$$
P_t = P_{t-1}\left(1+\frac{\pi_t}{100}\right)
$$

Dabei ist $P_t$ der durchschnittliche Preisindex im Jahr $t$ und $\pi_t$
die Inflationsrate in Prozent. Die Werte für 2027 und 2028 berechnen wir
in der nächsten Zelle.

Wir setzen den durchschnittlichen Preisindex für **2026 auf 100**. Für jedes
weitere Jahr erhöhen wir den Index um die prognostizierte Inflationsrate.

Für 2027 rechnen wir:

**100 × (1 + 2,7 / 100) = 102,7**

Für 2028 verwenden wir dann den Index von 2027 als Ausgangswert und erhöhen
ihn um 1,9 %. Den berechneten Wert schauen wir uns gemeinsam in der
Python-Ausgabe an.

Die Werte beziehen sich auf **Jahresdurchschnitte**, nicht auf den Preisstand
am 31. Dezember. Die Festlegung „2026 = 100“ dient nur als Vergleichsbasis.

```python
inflationspfad["inflation_dezimal"] = (
    inflationspfad["inflation_prozent"] / 100
)

inflationspfad["preisindex_jahresmittel_2026_100"] = (
    100 * (1 + inflationspfad["inflation_dezimal"]).cumprod()
)

inflationspfad.round(3)
```

### Einordnung und nächster Schritt

Damit sind Quelle, Veröffentlichungsstand, Preismaß und Planungshorizont
der externen Inflationsannahme nachvollziehbar festgelegt. Eine sinkende
Inflationsrate bedeutet zunächst, dass die Preise langsamer steigen;
sie bedeutet nicht automatisch einen Rückgang des Preisniveaus.

Der Preisindex wird nach Ausführung der Zellen gemeinsam geprüft und
interpretiert. Anschließend definieren wir das Beispielunternehmen:
Investitionsbetrag, Zahlungsüberschüsse, verglichene Rechtsformen,
Finanzierungsalternativen und steuerliche Annahmen. Erst danach
berechnen wir die finanziellen Folgen der Szenarien.

### Sechsjähriger Planungshorizont

Der bisher berechnete Preisindex enthält die beiden Jahre, für die wir
Werte aus der Bundesbank-Prognose übernommen haben. Für die
Finanzierungsentscheidung betrachten wir nun insgesamt sechs Jahre.

Wir verwenden drei Inflationspfade:

| Szenario | Festlegung |
|---|---|
| Basis | Bundesbank-Werte für 2027–2028; für 2029–2032 technisch angenommene 2,0 % pro Jahr |
| Höhere Inflation | In jedem Planjahr ein Prozentpunkt über dem Basispfad |
| Niedrigere Inflation | In jedem Planjahr ein Prozentpunkt unter dem Basispfad |

Die Zu- und Abschläge sind **illustrative Sensitivitätsannahmen**. Wir
behaupten damit weder, diese Verläufe vorhergesagt zu haben, noch weisen
wir ihnen Eintrittswahrscheinlichkeiten zu. Insbesondere ist der
Basispfad ab 2029 eine Modellannahme.

Zunächst verändern wir nur den Inflationspfad. Wie sich ein zusätzlicher
Energiepreisschock auf Unternehmenskosten und Absatz auswirken könnte,
untersuchen wir später im Unternehmensmodell. Auch Kreditzinsen leiten
wir nicht automatisch aus der Inflationsrate ab.

## 3. Unternehmen, Rechtsformen und Finanzierungsalternativen

### 3.1 Was wird entschieden?

Wir betrachten eine Investition eines fiktiven Unternehmens in Deutschland.
Für dieselbe Investition vergleichen wir mögliche Rechtsformen und Wege, das
benötigte Kapital bereitzustellen.

Wir unterscheiden fünf Finanzierungswege:

| Kürzel | Finanzierung in Betracht |
|---|---|
| F1 | Selbstfinanzierung aus Mitteln des Unternehmens |
| F2 | Zusätzliche Beteiligung der bisherigen Eigentümer |
| F3 | Beteiligung neuer Eigentümer |
| F4 | Darlehen der bisherigen Eigentümer |
| F5 | Fremdfinanzierung durch Dritte |

Wir modellieren für Kapital- und Personengesellschaften.
Ein Einzelunternehmen nehmen wir als zusätzliche Rechtsform in unseren
Untersuchungsraum auf; seine steuerliche Behandlung müssen wir eigenständig
und mit angegebenem Rechtsstand festlegen.

Für F5 ergänzen wir eine Vertragsfrage aus unserem Beratungsfall: Das
Darlehen eines externen Kreditgebers kann einen festen oder einen variablen
Zinssatz haben.

F3 nehmen wir in den ersten rechnerischen Vergleich noch nicht auf:
Neue Eigentümer würden Anteile am zukünftigen Vermögen erhalten. Dafür
müssten wir ihre Einlage, ihre Beteiligungsquote und die Perspektive der
bisherigen Eigentümer zusätzlich definieren. F3 bleibt als Erweiterung
sichtbar.

### 3.2 Wann ist der Vergleich fair?

Alle Alternativen beziehen sich auf dieselbe Investition, denselben
Planungshorizont und innerhalb eines Szenarios auf dieselbe geschäftliche
Entwicklung. Wir betrachten das Vermögen des Unternehmens und der
bisherigen Eigentümer zusammen.

Das ist bei den Finanzierungswegen entscheidend:

- Wird Kapital aus dem Unternehmen verwendet, steht es dort nicht mehr
  für eine andere Verwendung zur Verfügung.
- Bringen Eigentümer privates Geld ein, können sie dieses Geld nicht
  gleichzeitig außerhalb des Unternehmens anlegen.
- Bei einem Darlehen der Eigentümer sind Zinsen und Tilgung Zahlungen
  zwischen Unternehmen und Eigentümern. Wir dürfen sie beim gemeinsamen
  Vermögen nicht doppelt zählen; ihre steuerliche Behandlung kann sich
  dennoch auswirken.
- Bei einem Bankdarlehen fließen Zinsen und Tilgung an einen externen
  Kreditgeber. Nicht verwendete eigene Mittel bleiben für andere
  Verwendungen verfügbar.

Wir vergleichen deshalb das **absolute Vermögen der Eigentümer nach
Steuern** am Ende des gleichen Zeitraums und zusätzlich die Liquidität
in jedem Planjahr. Die Konditionen der Darlehen und die verfügbaren
Eigenmittel müssen für jede Alternative ausdrücklich angegeben werden.

Der verwendete Preisindex allein kann die Reihenfolge der Alternativen
innerhalb desselben Szenarios nicht ändern: Alle Endvermögen würden durch
denselben positiven Wert geteilt. Inflation wird für die Entscheidung
erst dann relevant, wenn sie beispielsweise Kosten, Verkaufspreise,
vertragliche Zinsen oder steuerlich relevante nominale Beträge
unterschiedlich beeinflusst.

Die Steuerrechnungen aus dem Lehrbuch übernehmen wir nicht als geltendes
Recht für 2027. Das Buch liefert die Vergleichslogik; konkrete
Steuerannahmen werden später mit ihrem jeweiligen Rechtsstand ausgewiesen.

```python
finanzierungswege = {
    "F1": "Einbehaltene Mittel",
    "F2": "Zusätzliche Mittel bisheriger Eigentümer",
    "F4": "Darlehen bisheriger Eigentümer",
    "F5_fest": "Externes Darlehen mit festem Zins",
    "F5_variabel": "Externes Darlehen mit variablem Zins",
}

moegliche_wege = {
    "Einzelunternehmen": ["F1", "F2", "F5_fest", "F5_variabel"],
    "Personengesellschaft": [
        "F1", "F2", "F4", "F5_fest", "F5_variabel"
    ],
    "Kapitalgesellschaft": [
        "F1", "F2", "F4", "F5_fest", "F5_variabel"
    ],
}

zeilen = []

for rechtsform, wege in moegliche_wege.items():
    for kuerzel in wege:
        zeilen.append({
            "rechtsform": rechtsform,
            "finanzierungsweg": kuerzel,
            "beschreibung": finanzierungswege[kuerzel],
        })

alternativen = pd.DataFrame(zeilen)

alternativen
```

### 3.3 Einordnung der Alternativen

Die Tabelle zeigt die Kombinationen, die wir im Modell zunächst betrachten.
Sie enthält noch keine Rangfolge und keine Aussage, dass jede Finanzierung
einem konkreten Unternehmen tatsächlich angeboten würde.

F1 bis F5 stammen als Finanzierungskategorien aus dem Lehrbuch. Die
Aufteilung von F5 in feste und variable Zinsen sowie die Betrachtung des
Einzelunternehmens sind Erweiterungen unseres Beratungsfalls.

Im nächsten Schritt geben wir dem fiktiven Unternehmen konkrete Zahlen:
Investitionsbetrag, verfügbare eigene Mittel, betriebliche Zahlungsströme
und angebotene Kreditkonditionen. Erst danach können wir Steuern und
finanzielle Ergebnisse berechnen.

```python
basisraten = (
    inflationspfad
    .set_index("jahr")["inflation_prozent"]
    .to_dict()
)

# Ab 2029: ausdrücklich angenommene Werte, keine Bundesbank-Prognose.
basisraten.update({
    jahr: 2.0
    for jahr in range(2029, 2033)
})

abweichungen = {
    "Basis": 0.0,
    "Höhere Inflation (+1 PP)": 1.0,
    "Niedrigere Inflation (-1 PP)": -1.0,
}

zeilen = []

for szenario, abweichung in abweichungen.items():
    for jahr in range(2027, 2033):
        amtlicher_basispunkt = (
            szenario == "Basis" and jahr <= 2028
        )

        zeilen.append({
            "szenario": szenario,
            "jahr": jahr,
            "inflation_prozent": basisraten[jahr] + abweichung,
            "herkunft": (
                "Bundesbank-Prognose"
                if amtlicher_basispunkt
                else "Modellannahme"
            ),
        })

szenarien = pd.DataFrame(zeilen)

jahresfaktor = 1 + szenarien["inflation_prozent"] / 100

szenarien["preisindex_2026_100"] = (
    100
    * jahresfaktor.groupby(szenarien["szenario"]).cumprod()
)

szenarien.round(2)
```

```python
import matplotlib.pyplot as plt

farben = {
    "Basis": "#20639B",
    "Höhere Inflation (+1 PP)": "#D1495B",
    "Niedrigere Inflation (-1 PP)": "#2A9D8F",
}

fig, achsen = plt.subplots(1, 2, figsize=(12, 4.5))

for name, daten in szenarien.groupby("szenario", sort=False):
    for achse, spalte in zip(
        achsen,
        ["inflation_prozent", "preisindex_2026_100"]
    ):
        achse.plot(
            daten["jahr"],
            daten[spalte],
            marker="o",
            linewidth=2,
            color=farben[name],
            label=name,
        )

for achse in achsen:
    achse.axvspan(2028.5, 2032.3, color="#EAF0F6", alpha=0.7)
    achse.set_xticks(range(2027, 2033))
    achse.set_xlabel("Jahr")
    achse.grid(alpha=0.2)

bundesbank_punkte = szenarien[
    (szenarien["szenario"] == "Basis")
    & (szenarien["jahr"] <= 2028)
]

achsen[0].scatter(
    bundesbank_punkte["jahr"],
    bundesbank_punkte["inflation_prozent"],
    marker="s",
    s=72,
    facecolor="white",
    edgecolor=farben["Basis"],
    linewidth=2,
    zorder=5,
    label="Bundesbank: Basis 2027–2028",
)

achsen[0].set_title("Inflationsraten")
achsen[0].set_ylabel("Prozent pro Jahr")
achsen[0].legend(fontsize=7)

achsen[1].set_title("Durchschnittlicher Preisindex")
achsen[1].set_ylabel("Index: 2026 = 100")

fig.text(
    0.01,
    0.01,
    "Nur Basis 2027–2028: Bundesbank-Prognose vom 12.06.2026. "
    "Übrige Werte: Modellannahmen.",
    fontsize=8,
)

fig.tight_layout(rect=(0, 0.06, 1, 1))
plt.show()
```

### 3.4 Annahmen, Begründungen und offene Modellentscheidungen

Eine Szenarioanalyse ist nur dann für eine Beratung brauchbar, wenn ihre
Ergebnisse auf nachvollziehbaren Annahmen beruhen. Wir unterscheiden
deshalb drei Arten von Eingaben:

1. **Übernommene Informationen:** Angaben aus einer benannten Quelle.
2. **Modellentscheidungen:** Vereinfachungen, die den Vergleich ermöglichen.
3. **Offene Eingaben:** Werte, die erst nach Prüfung oder im Gespräch mit
   dem Unternehmen festgelegt werden können.

Das Lehrbuch von König und Wosnitza liefert die Systematik der
Finanzierungswege F1 bis F5 und untersucht insbesondere Kapital- und
Personengesellschaften. Seine Steuerregeln und Zahlenbeispiele übernehmen
wir nicht ungeprüft für eine Entscheidung im Jahr 2026. Die Betrachtung
des Einzelunternehmens und die Unterscheidung zwischen festem und variablem
externem Kredit sind Erweiterungen unseres Modells.

Für 2027 und 2028 verwenden wir die genannten Werte aus der
Bundesbank-Prognose. Die jährlichen 2,0 % für 2029 bis 2032 dienen
lediglich als **konstanter Rechenpfad**, weil uns für diese Jahre in der
verwendeten Quelle keine entsprechenden Werte vorliegen. Sie sind keine
amtliche Prognose. Auch die Abweichungen von jeweils einem Prozentpunkt
sind frei gewählte **Stressparameter**: Sie zeigen, wie empfindlich das
spätere Ergebnis auf einen veränderten Inflationspfad reagiert. Ob gerade
diese Abweichung für die Entscheidung ausreicht, prüfen wir später mit
weiteren Werten.

Alle drei Pfade sind zunächst fest vorgegebene Szenarien. Wir ordnen ihnen
keine Eintrittswahrscheinlichkeiten zu. Deshalb brauchen wir in diesem
Projekt weder einen Zufallszahlengenerator noch eine angenommene
Wahrscheinlichkeitsverteilung.

#### Was wir für das Unternehmen noch benötigen

Vor einer Ergebnisrechnung müssen wir die folgenden Größen festlegen und
begründen:

- **Investition:** Betrag, Zeitpunkt und wirtschaftliche Nutzungsdauer.
  Ein Unternehmensangebot oder ein eindeutig als fiktiv gekennzeichnetes
  Fallbeispiel wäre dafür eine mögliche Grundlage.
- **Geschäftliche Zahlungsströme:** erwartete Einnahmen und Ausgaben.
  Wir müssen begründen, ob und wie Verkaufspreise, Kosten und Absatzmengen
  auf ein Szenario reagieren.
- **Eigene Mittel:** verfügbares Kapital des Unternehmens und der
  bisherigen Eigentümer. Dadurch wird bestimmt, welche
  Finanzierungswege überhaupt durchführbar sind.
- **Kreditbedingungen:** angebotene Zinssätze, Zinsbindung, Tilgung,
  mögliche Anpassung eines variablen Zinssatzes und weitere Kosten.
  Einen Kreditzins leiten wir nicht allein aus der Inflationsrate ab.
- **Steuern:** maßgebliche Steuerregeln mit dokumentiertem Rechtsstand
  und ausdrücklich vereinfachten Annahmen, falls ein vollständiger
  Steuerfall für das Notebook zu umfangreich wäre.
- **Vergleichsmaßstab:** Vermögen der bisherigen Eigentümer nach Steuern
  am Ende des Planungshorizonts und verfügbare Liquidität in jedem Jahr.
  Zusätzlich müssen wir die anderweitige Verwendung eigener Mittel
  berücksichtigen.

Bei jeder dieser Größen prüfen wir später, ob eine Änderung der Annahme
die Rangfolge der Alternativen verändert. Besonders wichtig sind dabei
Annahmen, die den Vergleich zwischen Rechtsformen oder zwischen festem
und variablem Kredit unmittelbar beeinflussen.

#### Grenze der gegenwärtigen Szenarien

Die Inflationspfade beschreiben bislang nur das allgemeine Preisniveau.
Sie erzeugen für sich allein noch unterschiedliche Gewinne, Steuerzahlungen
oder Kreditkosten. Erst das Unternehmensmodell legt fest, welche Zahlungen
sich mit Inflation ändern und welche nicht. Eine Empfehlung aus der
bisherigen Grafik wäre daher verfrüht.

Eine mögliche spätere Erweiterung wäre eine stochastische Simulation.
Dafür müssten wir für jede Zufallsgröße begründen, welche Werte möglich
sind, welche Daten ihre Verteilung und Parameter stützen und welche
Abhängigkeiten zwischen den Größen bestehen. Diese Entscheidungen treffen
wir erst gemeinsam, falls wir eine solche Erweiterung durchführen.

### 3.5 Ausgangszahlungen der Investition

Für einen ersten, nachvollziehbaren Rechenschritt übernehmen wir die
Ausgangszahlen des Investitionsbeispiels aus König und Wosnitza,
Kapitel IV.2.7, S. 223. Kapitel IV.3.8 verwendet dieselbe
Ausgangskonstellation beim Vergleich mit der Personengesellschaft.

| Größe | Wert | Herkunft |
|---|---:|---|
| Investitionsausgabe zu Beginn | 100.000 Geldeinheiten | Lehrbuchbeispiel |
| Jährlicher Einzahlungsüberschuss | 26.983,63 Geldeinheiten | Lehrbuchbeispiel |
| Anzahl der Jahre mit Überschuss | 6 | Lehrbuchbeispiel |

Wir behandeln die Investitionsausgabe als Zahlung **vor dem ersten
Planjahr**, also am Ende von 2026. Die sechs jährlichen Überschüsse
fallen von 2027 bis 2032 an. Diese zeitliche Zuordnung ist unsere
Anpassung des Buchbeispiels an den Planungshorizont des Notebooks.

Im Buch ist der jährliche Überschuss konstant. Wir übernehmen ihn
zunächst als **nominal konstanten Betrag**. Das bedeutet: In jedem
Szenario fließt derselbe Geldbetrag zu. Die Inflationsrate verändert
in dieser ersten Rechnung nur dessen Kaufkraft.

Um einen Jahresbetrag in Geldeinheiten mit Kaufkraft des Jahres 2026
auszudrücken, teilen wir ihn durch den Preisindex des betreffenden
Jahres und multiplizieren mit 100:

$$
\text{Realer Überschuss}_t
=
\text{Nominaler Überschuss}_t
\cdot \frac{100}{\text{Preisindex}_t}.
$$

**Grenze dieser Annahme:** Ein tatsächliches Unternehmen könnte
Verkaufspreise erhöhen, andere Kosten tragen oder andere Mengen
verkaufen. Dafür enthält der einzelne Überschuss aus dem Lehrbuch
keine getrennten Angaben. Wir unterstellen solche Wirkungen daher
noch nicht. Ebenso berechnen wir hier weder Steuern noch Zinsen.

Die Buchwerte dienen als **Referenzfall zum Aufbau und zur Prüfung des
Modells**. Sie dürfen nicht als aktuelle Marktdaten oder als
prognostizierte Einnahmen eines realen Unternehmens gelesen werden.

```python
# Ausgangswerte aus König/Wosnitza, Kap. IV.2.7, S. 223.
investition = 100_000.00
ueberschuss_nominal = 26_983.63

zahlungsstroeme = szenarien[
    ["szenario", "jahr", "preisindex_2026_100"]
].copy()

zahlungsstroeme["ueberschuss_nominal"] = ueberschuss_nominal

zahlungsstroeme["ueberschuss_real_2026"] = (
    zahlungsstroeme["ueberschuss_nominal"]
    * 100
    / zahlungsstroeme["preisindex_2026_100"]
)

zahlungsstroeme.round(2)
```

### Interpretation: nominaler Betrag und Kaufkraft

Der angenommene jährliche Einzahlungsüberschuss beträgt in allen drei
Szenarien 26.983,63 Geldeinheiten. Weil wir ihn zunächst nominal
konstant halten, sinkt seine Kaufkraft bei steigenden Preisen.

Im Basisszenario entspricht der Überschuss im Jahr 2027 rund
26.274 Geldeinheiten mit Kaufkraft von 2026. Für 2032 sind es noch
rund 23.821. Bei höherer Inflation liegt der reale Wert im Jahr 2032
bei rund 22.468, bei niedrigerer Inflation bei rund 25.270.

Die Unterschiede folgen vollständig aus unseren Inflationspfaden und
der Annahme eines nominal konstanten Überschusses. Sie belegen nicht,
dass die Investition in einem Szenario tatsächlich weniger Geld
einbringen würde: Dazu müssten wir Einnahmen, Kosten und Absatz
getrennt modellieren.

Aus dieser Tabelle allein lässt sich noch keine Finanzierung
empfehlen. Steuern, Zinsen, Tilgung und die Verwendung eigener Mittel
sind noch nicht berücksichtigt. Wir addieren die Jahresbeträge auch
noch nicht zu einem Entscheidungswert, solange der Vergleichsmaßstab
für Zahlungen zu verschiedenen Zeitpunkten nicht festgelegt ist.

```python
fig, ax = plt.subplots(figsize=(9, 5))

for name, daten in zahlungsstroeme.groupby("szenario", sort=False):
    ax.plot(
        daten["jahr"],
        daten["ueberschuss_real_2026"],
        marker="o",
        linewidth=2,
        color=farben[name],
        label=name,
    )

ax.scatter(
    2026,
    ueberschuss_nominal,
    marker="D",
    s=65,
    color="#555555",
    label="Buchbetrag als Referenz für 2026",
)

ax.set(
    title="Kaufkraft des Investitionsüberschusses",
    xlabel="Jahr",
    ylabel="Geldeinheiten mit Kaufkraft von 2026",
)
ax.set_xticks(range(2026, 2033))
ax.grid(alpha=0.2)
ax.legend()

fig.tight_layout()
plt.show()
```

### 3.6 Referenzrechnung: Externes Fälligkeitsdarlehen (F5)

Wir betrachten zunächst nur die externe Fremdfinanzierung F5 als
Referenzrechnung. Im Beispiel von König und Wosnitza, Kapitel IV.2.7,
S. 223, beträgt der angenommene Kreditzins 12 %. Als Kreditform wird
ein Fälligkeitsdarlehen verwendet.

Für die Rechnung setzen wir voraus, dass die Investitionsausgabe von
100.000 Geldeinheiten zu Beginn vollständig durch dieses Darlehen
finanziert wird. Jährlich fallen Zinsen von 12 % auf die unveränderte
Darlehenssumme an. Die vollständige Tilgung erfolgt am Ende des
sechsten Planjahres.

Die 12 % stammen aus dem historischen Lehrbuchbeispiel. Sie stellen
**kein aktuelles Kreditangebot** dar. Für unseren zusätzlichen
Vergleich zwischen festem und variablem Zins müssen wir später
gesonderte Vertragsbedingungen festlegen. Einen variablen Zinspfad
leiten wir insbesondere nicht automatisch aus der Inflationsrate ab.

Wir berechnen jetzt ausschließlich die Zahlungen für Investition,
Zinsen und Tilgung **vor Steuern**. Ein negativer Betrag im letzten
Jahr bedeutet für sich genommen noch keine Zahlungsunfähigkeit:
Dafür müssten wir auch vorhandene Mittel und zurückgelegte
Überschüsse kennen.

```python
kreditbetrag = investition
zins_fest_buch = 0.12
letztes_planjahr = 2032

f5_referenz = zahlungsstroeme[
    ["szenario", "jahr", "ueberschuss_nominal"]
].copy()

f5_referenz["zinsen"] = kreditbetrag * zins_fest_buch

f5_referenz["tilgung"] = (
    f5_referenz["jahr"]
    .eq(letztes_planjahr)
    .astype(int)
    * kreditbetrag
)

f5_referenz["zahlung_nach_kreditdienst_vor_steuern"] = (
    f5_referenz["ueberschuss_nominal"]
    - f5_referenz["zinsen"]
    - f5_referenz["tilgung"]
)

f5_referenz.round(2)
```

### 3.7 Rückzahlungsreserve für das Fälligkeitsdarlehen

Die Zahlung im Jahr 2032 ist für sich genommen negativ, weil dann die
gesamte Darlehenssumme fällig wird. Für die Beurteilung der Rückzahlung
betrachten wir deshalb zusätzlich die Überschüsse aus den Vorjahren.

In dieser Referenzrechnung wird jeder Überschuss nach Kreditzinsen
vollständig zurückgelegt. Er steht nicht für Ausschüttungen oder
andere Ausgaben zur Verfügung. Das zu Beginn eines Jahres vorhandene
Guthaben wird verzinst; der neue Jahresüberschuss fließt erst am
Jahresende in die Reserve.

König und Wosnitza nehmen in ihrem Beispiel einen Guthabenzins von
10 % an (Kapitel IV.2.7, S. 223). Wir verwenden diesen Wert hier als
**historische Buchannahme**, nicht als aktuell erzielbare Verzinsung.
Die Tilgung von 100.000 Geldeinheiten erfolgt am Ende von 2032.

Die Rechnung bleibt vor Steuern. Insbesondere berücksichtigen wir
noch keine Besteuerung der Guthabenzinsen. Die Reserve ist außerdem
nur ein abgegrenzter Teil des Unternehmensvermögens. Ihr Verlauf
beschreibt nicht die gesamte Unternehmensliquidität.

Da Überschüsse, Kreditzins und Guthabenzins bisher in allen
Inflationsszenarien nominal gleich sind, genügt zunächst eine
nominale Referenzrechnung. Anschließend können wir den Endbetrag
mit den jeweiligen Preisindizes in Kaufkraft von 2026 umrechnen.

```python
guthabenzins_buch = 0.10
reserve = 0.0
reserve_zeilen = []

f5_basis = f5_referenz[
    f5_referenz["szenario"] == "Basis"
].sort_values("jahr")

for _, jahr in f5_basis.iterrows():
    anfangsreserve = reserve
    guthabenzinsen = anfangsreserve * guthabenzins_buch

    reserve = (
        anfangsreserve
        + guthabenzinsen
        + jahr["zahlung_nach_kreditdienst_vor_steuern"]
    )

    reserve_zeilen.append({
        "jahr": int(jahr["jahr"]),
        "reserve_jahresanfang": anfangsreserve,
        "guthabenzinsen": guthabenzinsen,
        "zahlung_nach_kreditdienst": (
            jahr["zahlung_nach_kreditdienst_vor_steuern"]
        ),
        "reserve_jahresende": reserve,
    })

reserve_verlauf = pd.DataFrame(reserve_zeilen)

reserve_verlauf.round(2)
```

### 3.8 Wie empfindlich ist die Rückzahlung gegenüber dem Guthabenzins?

Mit dem Guthabenzins von 10 % aus dem Lehrbuch verbleibt nach der
Tilgung eine positive Reserve. Ob das auch bei geringerer Verzinsung
gilt, ist eine eigene Frage.

Wir berechnen deshalb zunächst zwei Referenzfälle:

- **0 %:** mathematischer Vergleichsfall ohne Guthabenzinsen;
  keine Prognose für ein Anlageangebot.
- **10 %:** historische Annahme des Lehrbuchbeispiels;
  ebenfalls keine aktuelle Marktprognose.

Anschließend suchen wir rechnerisch den Guthabenzins, bei dem die
Reserve Ende 2032 genau null wäre. Dieser **Schwellenwert** beschreibt
die Grenze innerhalb unseres vereinfachten Modells. Er ist weder ein
erwarteter Zinssatz noch eine Empfehlung für eine Geldanlage.

Die Rechnung erfolgt weiterhin vor Steuern. Eine positive Reserve
beweist noch nicht, dass das gesamte Unternehmen jederzeit liquide
ist: Andere Zahlungen, Ausschüttungen und Steuern fehlen bislang.

```python
def endreserve(guthabenzins):
    stand = 0.0

    for zahlung in f5_basis[
        "zahlung_nach_kreditdienst_vor_steuern"
    ]:
        stand = stand * (1 + guthabenzins) + zahlung

    return stand


untergrenze = 0.0
obergrenze = guthabenzins_buch

# Halbierung des Intervalls bis zum Guthabenzins mit Endreserve 0.
for _ in range(40):
    mitte = (untergrenze + obergrenze) / 2

    if endreserve(mitte) < 0:
        untergrenze = mitte
    else:
        obergrenze = mitte

schwellenzins = (untergrenze + obergrenze) / 2

zinsvergleich = pd.DataFrame({
    "Fall": ["Ohne Verzinsung", "Schwellenwert", "Buchannahme"],
    "Guthabenzins_prozent": [
        0,
        100 * schwellenzins,
        100 * guthabenzins_buch,
    ],
    "Endreserve_vor_steuern": [
        endreserve(0),
        endreserve(schwellenzins),
        endreserve(guthabenzins_buch),
    ],
})

zinsvergleich.round(2)
```

### 3.9 Erster fairer Vergleich: Eigene Mittel (F1) und Bankdarlehen (F5)

Wir vergleichen zwei Möglichkeiten, dieselbe Investition von
100.000 Geldeinheiten zu finanzieren:

| Alternative | Verwendung der anfänglich verfügbaren eigenen Mittel |
|---|---|
| F1: Eigene Mittel | 100.000 werden für die Investition eingesetzt. |
| F5: Bankdarlehen | Der Kredit bezahlt die Investition; die eigenen 100.000 können angelegt werden. |

Für diesen Vergleich **setzen wir voraus**, dass zu Beginn
100.000 Geldeinheiten eigene, frei verfügbare Mittel vorhanden sind.
Das ist eine Bedingung des Vergleichs und keine Information über ein
tatsächliches Unternehmen. Falls die Mittel fehlen, ist F1 in dieser
Form nicht durchführbar.

Wir verwenden weiter die historischen Beispielwerte des Buchs:
10 % Guthabenzins, 12 % Kreditzins und den jährlichen
Investitionsüberschuss von 26.983,63 Geldeinheiten. Jährliche freie
Überschüsse werden jeweils am Jahresende zurückgelegt. Bei F5 wird
die Darlehenssumme Ende 2032 getilgt.

Für beide Alternativen betrachten wir das am Ende vorhandene
Vermögen aus dieser Entscheidung:

- Bei F1 ist es die angesparte Summe der Investitionsüberschüsse.
- Bei F5 besteht es aus der Reserve nach Zinsen und Tilgung **plus**
  dem Endwert der anfänglich nicht eingesetzten eigenen Mittel.

Nur so berücksichtigen wir die anfänglichen 100.000 Geldeinheiten
in beiden Alternativen. Die Rechnung erfolgt vor Steuern und ist
noch keine Empfehlung für einen heutigen Kredit.

```python
eigene_mittel = investition
anzahl_jahre = len(f5_basis)

# F1: Eigene Mittel finanzieren die Investition.
reserve_f1 = 0.0

for _ in range(anzahl_jahre):
    reserve_f1 = (
        reserve_f1 * (1 + guthabenzins_buch)
        + ueberschuss_nominal
    )

# F5: Die eigenen Mittel bleiben angelegt.
reserve_f5 = reserve_verlauf.iloc[-1]["reserve_jahresende"]

endwert_eigene_mittel_f5 = (
    eigene_mittel
    * (1 + guthabenzins_buch) ** anzahl_jahre
)

vergleich_f1_f5 = pd.DataFrame({
    "Finanzierung": ["F1: Eigene Mittel", "F5: Bankdarlehen"],
    "Investitionsreserve_2032": [
        reserve_f1,
        reserve_f5,
    ],
    "Endwert_nicht_eingesetzter_Mittel": [
        0.0,
        endwert_eigene_mittel_f5,
    ],
})

vergleich_f1_f5["Endvermoegen_vor_Steuern"] = (
    vergleich_f1_f5["Investitionsreserve_2032"]
    + vergleich_f1_f5["Endwert_nicht_eingesetzter_Mittel"]
)

vergleich_f1_f5.round(2)
```

### 3.10 F1 und F5 unter den drei Inflationspfaden

Der bisherige Vergleich liefert Endwerte in nominalen Geldeinheiten
des Jahres 2032. Um ihre Kaufkraft zu vergleichen, rechnen wir beide
Endwerte mit dem Preisindex des jeweiligen Szenarios in Kaufkraft von
2026 um.

Dabei ist eine wichtige Modelleigenschaft zu beachten: Bisher sind
Investitionsüberschuss, Kredit- und Guthabenzins in allen drei
Inflationsszenarien nominal gleich. Innerhalb eines Szenarios teilen
wir daher die Endwerte von F1 und F5 durch **denselben positiven
Preisindex**.

Die Inflation kann so die Kaufkraft der Endwerte verändern. Sie kann
unter diesen Annahmen aber **nicht die Reihenfolge von F1 und F5
umkehren**. Das ist eine mathematische Folge unseres gegenwärtigen
Modells, keine allgemeine Aussage über Finanzierungsentscheidungen.

Für eine mögliche Änderung der Rangfolge müssten wir begründet
modellieren, wie Inflation beispielsweise Verkaufspreise, Kosten,
variable Kreditzinsen oder steuerliche Zahlungen unterschiedlich
beeinflusst. Diese Beziehungen sind mit den bislang verfügbaren
Angaben aus dem Buch noch nicht bestimmt.

```python
preisindex_2032 = szenarien.loc[
    szenarien["jahr"] == 2032,
    ["szenario", "preisindex_2026_100"]
]

vergleich_real = preisindex_2032.merge(
    vergleich_f1_f5[
        ["Finanzierung", "Endvermoegen_vor_Steuern"]
    ],
    how="cross",
)

vergleich_real["Endwert_mit_Kaufkraft_2026"] = (
    vergleich_real["Endvermoegen_vor_Steuern"]
    * 100
    / vergleich_real["preisindex_2026_100"]
)

vergleich_real[
    [
        "szenario",
        "Finanzierung",
        "Endvermoegen_vor_Steuern",
        "Endwert_mit_Kaufkraft_2026",
    ]
].round(2)
```

## 4. Steuerliche Modellierung: Rechtsstand und fehlende Bemessungsgrundlagen

### 4.1 Was können wir bereits aus einer Rechtsquelle übernehmen?

Für eine Kapitalgesellschaft gibt § 23 Körperschaftsteuergesetz
(KStG) die Körperschaftsteuersätze für unsere Planjahre vor:

| Jahr | Körperschaftsteuersatz |
|---|---:|
| 2027 | 15 % |
| 2028 | 14 % |
| 2029 | 13 % |
| 2030 | 12 % |
| 2031 | 11 % |
| 2032 | 10 % |

**Quelle und Rechtsstand:** [§ 23 KStG, Gesetze im Internet](
https://www.gesetze-im-internet.de/kstg_1977/__23.html),
abgerufen für dieses Modell am 24. September 2026.

Diese Tabelle enthält **Steuersätze, keine berechneten Steuern**.
Die Körperschaftsteuer knüpft an das zu versteuernde Einkommen an.
Unser bisheriger jährlicher Einzahlungsüberschuss aus dem
Lehrbuchbeispiel ist nicht automatisch dieses Einkommen. Für eine
Steuerrechnung fehlen insbesondere Annahmen zur steuerlichen
Gewinnermittlung und Abschreibung.

Hinzu kommen weitere Ebenen: Der Gewerbesteuer-Hebesatz hängt von
der Gemeinde ab. Bei Einzelunternehmen und Personengesellschaften
müssen wir die Besteuerung der Eigentümer betrachten. Bei einer
Kapitalgesellschaft ist zusätzlich relevant, ob Gewinne im
Unternehmen bleiben oder an Eigentümer ausgeschüttet werden.

Deshalb vergleichen wir noch keine Rechtsformen anhand einer
pauschalen „Steuerquote“. Wir dokumentieren erst die einschlägigen
Regeln und legen anschließend die fehlenden Angaben zum
Fallunternehmen fest. Auch die Geldeinheiten des Buchbeispiels
müssen ausdrücklich als Euro-Fall festgelegt werden, bevor wir
gesetzliche Freibeträge in Euro darauf anwenden.

```python
koerperschaftsteuer_saetze = pd.DataFrame({
    "jahr": range(2027, 2033),
    "kst_satz_prozent": [15, 14, 13, 12, 11, 10],
    "grundlage": "§ 23 KStG",
})

koerperschaftsteuer_saetze
```

### 4.2 Vom Einzahlungsüberschuss zum vereinfachten Teilergebnis

Der jährliche Einzahlungsüberschuss von 26.983,63 Geldeinheiten
beschreibt zunächst einen Zahlungsstrom. Für eine Steuerrechnung
benötigen wir dagegen eine passende Ergebnis- beziehungsweise
Einkommensgröße.

Das Lehrbuch legt für die Investition folgende Abschreibungen fest
(König und Wosnitza, Kapitel IV.2.7, S. 225):

| Jahr unseres Modells | Abschreibung nach dem Buchbeispiel |
|---|---:|
| 2027 | 20.000 |
| 2028–2032 | jeweils 16.000 |

Die Summe der sechs Abschreibungen beträgt 100.000 Geldeinheiten.
Die Investitionsausgabe erfolgt zu Beginn; die Abschreibungen
verteilen diesen Betrag im Buchbeispiel auf die Planjahre.
Eine Abschreibung ist in diesen Jahren **keine zusätzliche
Auszahlung**.

Für eine erste Rechenbrücke ziehen wir die Buchabschreibung vom
jährlichen Investitionsüberschuss ab. Beim externen Darlehen zeigen
wir zusätzlich die jährlichen Kreditzinsen getrennt. Die Tilgung
vermindert den Kreditbestand; wir ziehen sie in dieser
vereinfachten Ergebnisrechnung nicht noch einmal als Aufwand ab.

Die entstehenden Werte sind **Teilergebnisse dieser Investition
vor weiteren steuerlichen Anpassungen**. Sie sind ausdrücklich
noch nicht das zu versteuernde Einkommen des gesamten Unternehmens:
Das Buch beschreibt auch Erträge aus bereits vorhandenen
Investitionen. Zudem fehlen unter anderem die genaue Behandlung
der Zinserträge aus der Reserve und die Prüfung der für unseren
Planungszeitraum zulässigen Abschreibung.

Wir multiplizieren die Teilergebnisse deshalb noch nicht mit den
Körperschaftsteuersätzen.

```python
ergebnis_bruecke = f5_basis[
    ["jahr", "ueberschuss_nominal", "zinsen", "tilgung"]
].copy()

ergebnis_bruecke["afa_buch"] = (
    ergebnis_bruecke["jahr"]
    .map({2027: 20_000})
    .fillna(16_000)
)

ergebnis_bruecke["teilergebnis_f1"] = (
    ergebnis_bruecke["ueberschuss_nominal"]
    - ergebnis_bruecke["afa_buch"]
)

ergebnis_bruecke["teilergebnis_f5"] = (
    ergebnis_bruecke["teilergebnis_f1"]
    - ergebnis_bruecke["zinsen"]
)

ergebnis_bruecke[
    [
        "jahr",
        "ueberschuss_nominal",
        "afa_buch",
        "zinsen",
        "tilgung",
        "teilergebnis_f1",
        "teilergebnis_f5",
    ]
].round(2)
```

### 4.3 Die neue Investition im bestehenden Unternehmen

Bislang haben wir nur das Teilergebnis der neuen Investition
betrachtet. Im Lehrbuch existiert daneben ein bereits laufendes
Geschäft. Für dieses werden folgende Beispielwerte angegeben
(König und Wosnitza, Kapitel IV.2.7, S. 223–224):

| Bereits eingesetztes Kapital | Angenommene Rendite | Jährlicher Überschuss |
|---:|---:|---:|
| 1.900.000 | 11 % | 209.000 |
| 1.900.000 | 13 % | 247.000 |
| 1.900.000 | 15 % | 285.000 |

Wir verwenden zunächst **13 % als Referenzfall des Buchbeispiels**.
Die anderen beiden Werte bleiben für eine spätere
Sensitivitätsprüfung sichtbar. Die Renditen sind historische
Beispielannahmen und keine Prognosen für unser Planungsjahr 2027.

Laut Buch verursachen die vorhandenen Investitionen im betrachteten
Zeitraum keine weiteren Abschreibungen. Für eine vereinfachte
Ergebnisbrücke addieren wir daher ihren jährlichen Überschuss zum
Teilergebnis der neuen Investition.

Diese Summe ist weiterhin **keine fertige steuerliche
Bemessungsgrundlage**: Insbesondere die Zinserträge auf angesparte
Mittel, weitere steuerliche Anpassungen und die Ebene der
Eigentümer fehlen noch. Das vorhandene Geschäft zeigt aber, warum
ein negatives Teilergebnis der neuen Investition nicht automatisch
einen Verlust des gesamten Unternehmens bedeutet.

```python
kapital_alt_buch = 1_900_000
rendite_alt_buch = 0.13

ueberschuss_alt_buch = (
    kapital_alt_buch * rendite_alt_buch
)

ergebnis_unternehmen = ergebnis_bruecke[
    ["jahr", "teilergebnis_f1", "teilergebnis_f5"]
].copy()

ergebnis_unternehmen["ueberschuss_alt_buch"] = (
    ueberschuss_alt_buch
)

ergebnis_unternehmen["vereinfachtes_ergebnis_f1"] = (
    ergebnis_unternehmen["ueberschuss_alt_buch"]
    + ergebnis_unternehmen["teilergebnis_f1"]
)

ergebnis_unternehmen["vereinfachtes_ergebnis_f5"] = (
    ergebnis_unternehmen["ueberschuss_alt_buch"]
    + ergebnis_unternehmen["teilergebnis_f5"]
)

ergebnis_unternehmen.round(2)
```

### 4.4 Isolierter Effekt der Kreditzinsen auf die Körperschaftsteuer

Das vereinfachte Unternehmensergebnis ist bei F5 in jedem Planjahr
um 12.000 Geldeinheiten geringer als bei F1. Dieser Unterschied
entspricht den jährlichen Kreditzinsen des Buchbeispiels.

Für eine **bedingte Teilrechnung** nehmen wir an, dass

1. die Kreditzinsen in diesem Fall vollständig abzugsfähig sind,
2. alle übrigen steuerlich relevanten Größen bei F1 und F5 gleich
   sind und
3. die steuerliche Bemessungsgrundlage in beiden Fällen positiv
   bleibt.

Unter genau diesen Bedingungen multiplizieren wir die
Ergebnisdifferenz mit dem Körperschaftsteuersatz des jeweiligen
Jahres. Das Ergebnis zeigt lediglich, wie stark **diese eine
Ergebnisdifferenz** die Körperschaftsteuer beeinflussen könnte.

Die Rechnung ist **keine ermittelte Steuerlast**: Das
Abschreibungsschema und die Geldbeträge stammen aus einem
historischen Buchbeispiel. Außerdem fehlen Zinserträge auf
angesparte Mittel, weitere steuerliche Anpassungen,
Gewerbesteuer, Solidaritätszuschlag und die Besteuerung der
Eigentümer.

Rechtsgrundlagen für die spätere Prüfung:
[§ 23 KStG](https://www.gesetze-im-internet.de/kstg_1977/__23.html),
[§ 4 Abs. 4 EStG](https://www.gesetze-im-internet.de/estg/__4.html)
und [§ 4h EStG](https://www.gesetze-im-internet.de/estg/__4h.html).

```python
kst_teileffekt = ergebnis_unternehmen[
    [
        "jahr",
        "vereinfachtes_ergebnis_f1",
        "vereinfachtes_ergebnis_f5",
    ]
].merge(
    koerperschaftsteuer_saetze[
        ["jahr", "kst_satz_prozent"]
    ],
    on="jahr",
)

kst_teileffekt["zinsbedingte_ergebnisdifferenz"] = (
    kst_teileffekt["vereinfachtes_ergebnis_f1"]
    - kst_teileffekt["vereinfachtes_ergebnis_f5"]
)

kst_teileffekt["bedingter_kst_teileffekt"] = (
    kst_teileffekt["zinsbedingte_ergebnisdifferenz"]
    * kst_teileffekt["kst_satz_prozent"]
    / 100
)

kst_teileffekt[
    [
        "jahr",
        "zinsbedingte_ergebnisdifferenz",
        "kst_satz_prozent",
        "bedingter_kst_teileffekt",
    ]
].round(2)
```

## 5. Entscheidungsschwelle für den Kreditzins

### 5.1 Wann sind F1 und F5 vor Steuern gleichwertig?

Unser bisheriger F1/F5-Vergleich verwendet dieselbe Investition,
dieselben jährlichen Investitionsüberschüsse und denselben
Guthabenzins für zurückgelegte Beträge. F5 unterscheidet sich
dadurch, dass die eigenen 100.000 Geldeinheiten angelegt werden
können, während auf den gleich hohen Kredit Zinsen anfallen.

Unter genau diesen Annahmen gilt für das Fälligkeitsdarlehen:
**F1 und F5 sind vor Steuern gleichwertig, wenn Kreditzins und
Guthabenzins gleich hoch sind.**

Im historischen Buchbeispiel beträgt der Guthabenzins 10 % und
der Kreditzins 12 %. Der jährliche Zinsabstand von zwei
Prozentpunkten wirkt über die sechs Jahre auf das Endvermögen.
Die Investitionsüberschüsse sind in diesem Vergleich bei beiden
Alternativen gleich und bestimmen daher nicht den Unterschied
zwischen ihnen.

Diese Grenze gilt für das hier modellierte Darlehen und die
angenommene Wiederanlage der Zahlungen. Steuern, Gebühren,
unterschiedliche Vertragsbedingungen und variable Zinssätze
können die Grenze verändern. Wir verwenden sie deshalb als
**Vorsteuer-Referenz** für die weitere Analyse.

```python
# Endwertfaktor für sechs Zahlungen jeweils am Jahresende.
endwertfaktor = sum(
    (1 + guthabenzins_buch) ** k
    for k in range(anzahl_jahre)
)

gleichstands_zins = guthabenzins_buch

# Abstand der Endvermögen F1 minus F5 im Buchreferenzfall.
zinsbedingter_endwertabstand = (
    kreditbetrag
    * (zins_fest_buch - guthabenzins_buch)
    * endwertfaktor
)

zinsschwelle = pd.DataFrame({
    "Groesse": [
        "Guthabenzins im Buchbeispiel",
        "Kreditzins bei Gleichstand",
        "Kreditzins im Buchbeispiel",
        "Endwertfaktor der sechs Jahreszahlungen",
        "Vorsteuer-Endwertvorteil von F1 im Buchbeispiel",
    ],
    "Wert": [
        100 * guthabenzins_buch,
        100 * gleichstands_zins,
        100 * zins_fest_buch,
        endwertfaktor,
        zinsbedingter_endwertabstand,
    ],
})

zinsschwelle.round(2)
```

### 5.2 Bedingte Schwelle für einen später veränderlichen Kreditzins

Ein variabler Kredit benötigt eine vertragliche Regel für
Zinsanpassungen. Einen solchen Vertrag und einen zukünftigen
Zinspfad haben wir nicht vorliegen. Wir können aber eine
**Bedingungsfrage** an unser Modell stellen:

> Angenommen, der Kreditzins beträgt in den ersten beiden
> Planjahren 12 % wie im historischen Buchbeispiel. Wie hoch
> dürfte ein einheitlicher Kreditzins für die Jahre 2029–2032
> höchstens sein, damit F5 vor Steuern mit F1 gleichzieht?

Die Zweiteilung ist eine ausdrücklich gewählte Vereinfachung
für diese Schwellenwertanalyse, **kein Kreditangebot und keine
Zinsprognose**. Der noch unbekannte Zins für 2029–2032 wird
berechnet und nicht als zukünftiger Marktwert ausgegeben.

Ein Zins im frühen Planjahr wirkt länger auf das Endvermögen
als ein Zins im letzten Jahr. Deshalb gewichten wir jeden
jährlichen Zinsabstand zwischen Kredit- und Guthabenzins
mit der Anzahl der verbleibenden Anlagejahre.

Wir ändern hier nur die Kreditkondition. Aus einem
Inflationsszenario folgt kein bestimmter Kreditzins. Für
eine gemeinsame Betrachtung brauchen wir später entweder
konkrete Vertragsbedingungen oder ausdrücklich begründete
kombinierte Stressszenarien.

```python
planjahre = list(range(2027, 2033))
jahre_mit_buchzins = [2027, 2028]
jahre_mit_offenem_zins = [2029, 2030, 2031, 2032]

# Wert eines Zinsbetrags aus Jahr t am Ende von 2032.
gewicht = {
    jahr: (1 + guthabenzins_buch) ** (2032 - jahr)
    for jahr in planjahre
}

gewicht_buchzins = sum(
    gewicht[jahr] for jahr in jahre_mit_buchzins
)
gewicht_offener_zins = sum(
    gewicht[jahr] for jahr in jahre_mit_offenem_zins
)

# Gesucht: späterer Zins, bei dem der gewichtete
# Kredit-Zinsaufwand dem entgangenen Guthabenzins entspricht.
spaeterer_zins_bei_gleichstand = (
    guthabenzins_buch
    * (gewicht_buchzins + gewicht_offener_zins)
    - zins_fest_buch * gewicht_buchzins
) / gewicht_offener_zins

zinsgrenze_variabel = pd.DataFrame({
    "Groesse": [
        "Kreditzins 2027–2028: gesetzte Buchreferenz",
        "Gesuchter Kreditzins 2029–2032 bei Gleichstand",
        "Guthabenzins: gesetzte Buchreferenz",
    ],
    "Prozent_pro_Jahr": [
        100 * zins_fest_buch,
        100 * spaeterer_zins_bei_gleichstand,
        100 * guthabenzins_buch,
    ],
})

zinsgrenze_variabel.round(2)
```

### 5.3 Gemeinsame Sensitivität: Kreditzins und Inflation

Wir verbinden zwei bisher getrennte Fragen: Wie verändert ein späterer
Kreditzins das Endvermögen und die Rückzahlungsreserve? Welche Kaufkraft
haben die Ergebnisse unter den drei Inflationspfaden?

Für 2027–2028 bleibt der Kreditzins bei 12 % aus dem Lehrbuchbeispiel.
Für 2029–2032 vergleichen wir drei **bedingte** Zinssätze:

| Späterer Kreditzins | Begründung |
|---|---|
| Rechnerischer Gleichstand | Ergebnis aus Abschnitt 5.2 |
| 10 % | Guthabenzins des Lehrbuchbeispiels als Vergleichswert |
| 12 % | Unveränderter Kreditzins des Lehrbuchbeispiels |

Diese Werte sind weder Kreditangebote noch Prognosen. Wir kombinieren
jeden Zinsfall mit jedem bereits definierten Inflationspfad, ohne eine
statistische Abhängigkeit zwischen Inflation und Kreditzins zu behaupten.

Der jährliche Investitionsüberschuss und der Guthabenzins bleiben
nominal konstant. Deshalb verändert die Inflation hier die Kaufkraft
des Endvermögens, aber nicht dessen nominalen Betrag. Die
**Rückzahlungsreserve Ende 2032** zeigt, was nach Zinsen und Tilgung
aus den zurückgelegten Investitionsüberschüssen verbleibt. Sie ist
nicht mit der gesamten Liquidität des Unternehmens gleichzusetzen.

Alle Ergebnisse bleiben vor Steuern. Insbesondere sind die bisher
berechneten Körperschaftsteuer-Teileffekte noch keine vollständigen
Steuerzahlungen, die wir vom Endvermögen abziehen könnten.

```python
zinsfaelle = {
    "Gleichstand aus 5.2": spaeterer_zins_bei_gleichstand,
    "Guthabenzins des Buchs": guthabenzins_buch,
    "Unveränderter Buchzins": zins_fest_buch,
}

preisindex_am_ende = (
    szenarien.loc[
        szenarien["jahr"].eq(2032),
        ["szenario", "preisindex_2026_100"],
    ]
    .set_index("szenario")["preisindex_2026_100"]
)

ergebnisse = []

for zinsfall, spaeterer_zins in zinsfaelle.items():
    reserve_f5_szenario = 0.0

    for jahr in range(2027, 2033):
        kreditzins = (
            zins_fest_buch
            if jahr <= 2028
            else spaeterer_zins
        )

        tilgung = (
            kreditbetrag
            if jahr == 2032
            else 0.0
        )

        reserve_f5_szenario = (
            reserve_f5_szenario * (1 + guthabenzins_buch)
            + ueberschuss_nominal
            - kreditbetrag * kreditzins
            - tilgung
        )

    vermoegen_f5 = (
        reserve_f5_szenario
        + endwert_eigene_mittel_f5
    )

    for inflationsfall, preisindex in preisindex_am_ende.items():
        ergebnisse.append({
            "Inflationsszenario": inflationsfall,
            "Kreditzinsfall_2029_2032": zinsfall,
            "Kreditzins_ab_2029_prozent": 100 * spaeterer_zins,
            "Rueckzahlungsreserve_F5_2032": reserve_f5_szenario,
            "Endvermoegen_F1_nominal": reserve_f1,
            "Endvermoegen_F5_nominal": vermoegen_f5,
            "Vorteil_F1_real_2026": (
                (reserve_f1 - vermoegen_f5)
                * 100 / preisindex
            ),
        })

gemeinsame_sensitivitaet = pd.DataFrame(ergebnisse)

gemeinsame_sensitivitaet[
    [
        "Inflationsszenario",
        "Kreditzinsfall_2029_2032",
        "Kreditzins_ab_2029_prozent",
        "Rueckzahlungsreserve_F5_2032",
        "Vorteil_F1_real_2026",
    ]
].round(2)
```

### 5.4 Interpretation der Entscheidungsschwelle

Die Rückzahlungsreserve ist in allen untersuchten Zinsfällen positiv.
Unter den gesetzten Annahmen können die angesparten Überschüsse das
Fälligkeitsdarlehen am Ende des Planungshorizonts decken. Das ist
eine Aussage über diese abgegrenzte Reserve vor Steuern, nicht über
die gesamte Liquidität des Unternehmens.

Für die Wahl zwischen F1 und F5 betrachten wir zusätzlich das
Endvermögen einschließlich der bei F5 angelegten eigenen Mittel.
Der später geltende Kreditzins von rund 8,68 % markiert dabei den
Gleichstand. Oberhalb dieser Grenze ist F1 im betrachteten Modell
vorteilhafter.

Die folgende Grafik zeigt den Vorteil von F1 in Kaufkraft des
Jahres 2026. Eine Linie steht jeweils für einen Inflationspfad.
Die verbundenen Zinswerte sind eine mathematische Sensitivität
zwischen den betrachteten Konditionen; sie sind keine Vorhersage
künftiger Bankzinsen.

Die Linien schneiden die Nulllinie beim gleichen Kreditzins:
Inflation verändert hier die Kaufkraft des Unterschieds, aber nicht
die Entscheidungsschwelle. Für einen Einfluss der Inflation auf
die Finanzierungswahl müssten wir begründet festlegen, wie sie
beispielsweise die Einnahmen, Kosten oder Kreditkonditionen
des Unternehmens verändert.

## 6. Finanzierung durch die bisherigen Eigentümer: F2 und F4

Neben der Verwendung bereits vorhandener Unternehmensmittel (F1)
betrachten wir zwei weitere Wege aus der Systematik von König und
Wosnitza, Kapitel IV.1.1:

- **F2:** Die bisherigen Eigentümer stellen 100.000 Geldeinheiten
  zusätzlich als Eigenkapital bereit.
- **F4:** Die bisherigen Eigentümer leihen dem Unternehmen
  100.000 Geldeinheiten.

Wir betrachten Unternehmen und bisherige Eigentümer gemeinsam.
Zu Beginn stehen für jede dieser Alternativen **dieselben
100.000 Geldeinheiten aus dem Eigentümerkreis** zur Verfügung.
Das ist eine Vergleichsannahme, keine Aussage darüber, wo sich
das Geld bei einem konkreten Unternehmen heute befindet.
Der Investitionsüberschuss und der Anlagezins bleiben wie bisher.

Für F4 setzen wir beispielhaft einen jährlichen Darlehenszins
von 12 % und die Tilgung Ende 2032 an. Die 12 % übernehmen wir
hier allein als bereits verwendeten Rechenwert aus dem
historischen F5-Beispiel. Sie sind **keine ermittelte Kondition
für ein Gesellschafterdarlehen**. Die Eigentümer legen erhaltene
Zinsen jeweils am Jahresende zu den angenommenen 10 % an.

Wir zeigen bei F4 das Guthaben des Unternehmens und das Guthaben
der Eigentümer getrennt. Zins und Tilgung sind Zahlungen zwischen
beiden Seiten. Im gemeinsamen Vermögen zählen sie deshalb nicht
als Abfluss an einen Dritten.

Diese Rechnung erfolgt vor Steuern und ohne unterschiedliche
rechtliche Ansprüche auf das Unternehmensvermögen. Gerade
Besteuerung, Beteiligungsrechte und die Liquidität *innerhalb*
des Unternehmens können F1, F2 und F4 später unterscheiden.

```python
zins_gesellschafterdarlehen = zins_fest_buch

guthaben_unternehmen_f4 = 0.0
guthaben_eigentuemer_f4 = 0.0
f4_zeilen = []

for jahr in range(2027, 2033):
    zinszahlung = (
        investition * zins_gesellschafterdarlehen
    )
    rueckzahlung = (
        investition if jahr == 2032 else 0.0
    )

    # Unternehmensseite: Überschuss abzüglich Zahlungen
    # an die bisherigen Eigentümer.
    guthaben_unternehmen_f4 = (
        guthaben_unternehmen_f4
        * (1 + guthabenzins_buch)
        + ueberschuss_nominal
        - zinszahlung
        - rueckzahlung
    )

    # Eigentümerseite: erhaltene Zinsen und Rückzahlung.
    # Frühere Zahlungen werden mit 10 % weiter angelegt.
    guthaben_eigentuemer_f4 = (
        guthaben_eigentuemer_f4
        * (1 + guthabenzins_buch)
        + zinszahlung
        + rueckzahlung
    )

    f4_zeilen.append({
        "jahr": jahr,
        "guthaben_unternehmen_f4": guthaben_unternehmen_f4,
        "guthaben_eigentuemer_f4": guthaben_eigentuemer_f4,
        "gemeinsames_vermoegen_f4": (
            guthaben_unternehmen_f4
            + guthaben_eigentuemer_f4
        ),
    })

f4_verlauf = pd.DataFrame(f4_zeilen)

f4_verlauf.round(2)
```

```python
# F2: Die bisherigen Eigentümer stellen das Kapital
# für dieselbe Investition bereit. Vor Steuern gelten
# hier dieselben Investitionszahlungen wie bei F1.
endvermoegen_f2 = reserve_f1

vergleich_eigentuemerfinanzierung = pd.DataFrame({
    "Finanzierung": [
        "F1: vorhandene eigene Mittel",
        "F2: neues Eigenkapital bisheriger Eigentümer",
        "F4: Darlehen bisheriger Eigentümer",
    ],
    "Gemeinsames_Endvermoegen_vor_Steuern": [
        reserve_f1,
        endvermoegen_f2,
        (
            guthaben_unternehmen_f4
            + guthaben_eigentuemer_f4
        ),
    ],
})

vergleich_eigentuemerfinanzierung.round(2)
```

## 7. Steuerliche Überleitung: Unternehmen und Eigentümer

Vor Steuern haben F1, F2 und F4 im vereinfachten Modell dasselbe
gemeinsame Endvermögen. Steuerlich müssen wir jedoch festhalten,
**wer eine Zinszahlung leistet und wer sie erhält**.

Bei F1 und F2 fällt in unserer Rechnung kein Darlehenszins an.
Bei F4 zahlt das Unternehmen jährlich 12.000 Geldeinheiten an
die bisherigen Eigentümer. Bei F5 zahlt es denselben Betrag
an eine externe Bank.

Wir erweitern zunächst ausschließlich die **bedingte
Körperschaftsteuer-Teilrechnung** aus Abschnitt 4:

- Die vereinfachte Ergebnisgröße stammt weiterhin aus dem
  historischen Buchbeispiel.
- Wir unterstellen für diesen isolierten Vergleich einen
  vollständigen Zinsabzug und eine positive steuerliche
  Bemessungsgrundlage.
- Wir berechnen nur den möglichen Effekt der Zinsdifferenz
  auf die Körperschaftsteuer der Gesellschaft.

Bei F4 steht einer möglichen Steuerminderung der Gesellschaft
ein Zinsertrag der Eigentümer gegenüber. Dessen Besteuerung
berechnen wir hier noch nicht. Bei F5 erhält dagegen ein
externer Kreditgeber die Zinsen. Die beiden gleich hohen
Zinszahlungen dürfen daher nicht als gleich hoher Vorteil
für die bisherigen Eigentümer interpretiert werden.

Für eine Personengesellschaft dürfen wir diese
Körperschaftsteuerrechnung nicht übernehmen. Dort müssen
insbesondere die Zinszahlungen an Mitunternehmer und ihre
Gewinnanteile gemeinsam betrachtet werden.

Rechtsgrundlagen für die spätere Einordnung:
[§ 15 EStG](https://www.gesetze-im-internet.de/estg/__15.html),
[§ 20 EStG](https://www.gesetze-im-internet.de/estg/__20.html),
[§ 32d EStG](https://www.gesetze-im-internet.de/estg/__32d.html)
und [§ 8 KStG](https://www.gesetze-im-internet.de/kstg_1977/__8.html).

```python
steuer_ueberleitung = (
    ergebnis_unternehmen[
        ["jahr", "vereinfachtes_ergebnis_f1"]
    ]
    .merge(
        koerperschaftsteuer_saetze[
            ["jahr", "kst_satz_prozent"]
        ],
        on="jahr",
    )
)

jahreszins = (
    investition * zins_gesellschafterdarlehen
)

# Isolierte Ergebnisunterschiede gegenüber F1.
# F2: kein Zins; F4: Zins an Eigentümer;
# F5: Zins an externen Kreditgeber.
steuer_ueberleitung["zins_f2"] = 0.0
steuer_ueberleitung["zins_f4_an_eigentuemer"] = jahreszins
steuer_ueberleitung["zins_f5_an_bank"] = (
    kreditbetrag * zins_fest_buch
)

steuer_ueberleitung["bedingter_kst_teileffekt_f2"] = 0.0

steuer_ueberleitung["bedingter_kst_teileffekt_f4"] = (
    steuer_ueberleitung["zins_f4_an_eigentuemer"]
    * steuer_ueberleitung["kst_satz_prozent"]
    / 100
)

steuer_ueberleitung["bedingter_kst_teileffekt_f5"] = (
    steuer_ueberleitung["zins_f5_an_bank"]
    * steuer_ueberleitung["kst_satz_prozent"]
    / 100
)

steuer_ueberleitung[
    [
        "jahr",
        "kst_satz_prozent",
        "zins_f4_an_eigentuemer",
        "zins_f5_an_bank",
        "bedingter_kst_teileffekt_f2",
        "bedingter_kst_teileffekt_f4",
        "bedingter_kst_teileffekt_f5",
    ]
].round(2)
```

### 7.1 Gesellschafterdarlehen bei einer Personengesellschaft

Bei einer Personengesellschaft müssen wir zwischen dem Ergebnis
der Gesellschaft und den Einkünften ihrer Mitunternehmer
unterscheiden. Erhält ein Mitunternehmer eine Vergütung für
ein Darlehen an die Gesellschaft, ist diese nach § 15 Abs. 1
Satz 1 Nr. 2 EStG bei seinen gewerblichen Einkünften zu
berücksichtigen.

Für unser vereinfachtes Beispiel bedeutet das:

- **F4:** Die Gesellschaft zahlt jährlich 12.000
  Geldeinheiten Zinsen an einen bisherigen Mitunternehmer.
  Wir zeigen das Ergebnis der Gesellschaft nach Zinsen und
  die Darlehensvergütung des Mitunternehmers getrennt.
- **F5:** Die Gesellschaft zahlt die Zinsen an eine externe
  Bank. Bei den bisherigen Eigentümern entsteht daraus
  keine entsprechende Darlehensvergütung.

Wir verwenden weiterhin nur die Ergebnisbrücke aus dem
historischen Buchbeispiel. Die folgende Summe ist **keine
fertige Einkommensteuer-Bemessungsgrundlage**: Weitere
steuerliche Anpassungen, die Zuordnung auf einzelne
Mitunternehmer, deren sonstige Einkünfte und die
Gewerbesteuer sind nicht berechnet.

Quelle: [§ 15 Abs. 1 Satz 1 Nr. 2 EStG](
https://www.gesetze-im-internet.de/estg/__15.html).

```python
personengesellschaft_bruecke = ergebnis_unternehmen[
    [
        "jahr",
        "vereinfachtes_ergebnis_f1",
        "vereinfachtes_ergebnis_f5",
    ]
].copy()

# F4: derselbe Darlehenszins wie in unserer F5-Referenz,
# aber Empfänger ist ein bisheriger Mitunternehmer.
personengesellschaft_bruecke["ergebnis_gesellschaft_f4"] = (
    personengesellschaft_bruecke[
        "vereinfachtes_ergebnis_f1"
    ]
    - jahreszins
)

personengesellschaft_bruecke[
    "darlehensverguetung_mitunternehmer_f4"
] = jahreszins

personengesellschaft_bruecke[
    "summe_gesellschaft_und_mitunternehmer_f4"
] = (
    personengesellschaft_bruecke[
        "ergebnis_gesellschaft_f4"
    ]
    + personengesellschaft_bruecke[
        "darlehensverguetung_mitunternehmer_f4"
    ]
)

personengesellschaft_bruecke[
    [
        "jahr",
        "vereinfachtes_ergebnis_f1",
        "ergebnis_gesellschaft_f4",
        "darlehensverguetung_mitunternehmer_f4",
        "summe_gesellschaft_und_mitunternehmer_f4",
        "vereinfachtes_ergebnis_f5",
    ]
].round(2)
```

## 8. Geschäftsrisiko: Grenze für die Kreditrückzahlung

Bislang beträgt der jährliche Einzahlungsüberschuss der neuen
Investition konstant 26.983,63 Geldeinheiten. Für die Beratung
ist zusätzlich wichtig, wie stark dieser Betrag zurückgehen
dürfte, bevor die angesparte F5-Reserve Ende 2032 nicht mehr
zur vollständigen Rückzahlung ausreicht.

Wir berechnen für jeden bereits betrachteten Kreditzinsfall den
**kleinsten konstanten jährlichen Überschuss**, bei dem die
Reserve nach der Tilgung genau null beträgt. Die Differenz zum
Buchwert ist der rechnerische Spielraum für einen dauerhaften
Rückgang des Überschusses.

Dies ist eine bedingte Stressgrenze. Wir behaupten weder, dass
der Überschuss tatsächlich sinkt, noch weisen wir einem
Rückgang eine Wahrscheinlichkeit zu. Ohne getrennte Angaben
zu Umsatz, Absatz und Kosten können wir insbesondere nicht
aus einem Inflations- oder Kriegsszenario ableiten, wie stark
der Überschuss sinken würde.

Die Grenze betrifft nur die aus dieser Investition gebildete
Reserve vor Steuern. Das bestehende Geschäft und die
anfänglich verfügbaren eigenen Mittel werden hier nicht
zur Rückzahlung herangezogen.

```python
# Ein Überschuss aus Jahr t wird bis Ende 2032 angelegt.
zahlungsgewichte = {
    jahr: (1 + guthabenzins_buch) ** (2032 - jahr)
    for jahr in range(2027, 2033)
}

summe_gewichte = sum(zahlungsgewichte.values())
risikogrenzen = []

for zinsfall, spaeterer_zins in zinsfaelle.items():
    gewichtete_kreditzinsen = sum(
        kreditbetrag
        * (
            zins_fest_buch
            if jahr <= 2028
            else spaeterer_zins
        )
        * zahlungsgewichte[jahr]
        for jahr in range(2027, 2033)
    )

    # Bei dieser Höhe decken angesparte Überschüsse
    # genau die Kreditzinsen und die Tilgung.
    mindest_ueberschuss = (
        kreditbetrag + gewichtete_kreditzinsen
    ) / summe_gewichte

    spielraum = (
        ueberschuss_nominal - mindest_ueberschuss
    )

    risikogrenzen.append({
        "Kreditzinsfall_2029_2032": zinsfall,
        "Mindestueberschuss_pro_Jahr": mindest_ueberschuss,
        "Buchueberschuss_pro_Jahr": ueberschuss_nominal,
        "Spielraum_pro_Jahr": spielraum,
        "Maximaler_Rueckgang_prozent": (
            100 * spielraum / ueberschuss_nominal
        ),
        "Endreserve_beim_Buchueberschuss": (
            spielraum * summe_gewichte
        ),
    })

geschaeftsrisiko = pd.DataFrame(risikogrenzen)

geschaeftsrisiko.round(2)
```

## 11. Bedingte Körperschaftsteuerrechnung: F1 gegen F5

Die bisherige Steuerrechnung zeigte nur, welchen isolierten
Effekt die Kreditzinsen auf die Körperschaftsteuer haben
könnten. Für einen besseren Vergleich berücksichtigen wir
nun zusätzlich die Erträge aus angesparten Mitteln. Wir
betrachten dafür **eine Kapitalgesellschaft, die ihre
verfügbaren Mittel im Unternehmen behält**.

### Annahmen für diesen abgegrenzten Vergleich

| Größe | Festlegung und Begründung |
|---|---|
| Investition und Überschuss | 100.000 zu Beginn; jährlich 26.983,63 aus dem Buchbeispiel |
| Eigene Mittel zu Beginn | 100.000 befinden sich bei beiden Alternativen zunächst im Unternehmen |
| F1 | Das Unternehmen setzt diese Mittel für die Investition ein |
| F5 | Ein Bankkredit finanziert die Investition; die eigenen Mittel bleiben im Unternehmen angelegt |
| Kredit | 12 % jährlich und Tilgung Ende 2032 wie in der historischen Referenzrechnung |
| Guthaben | 10 % jährlich wie im Buchbeispiel; Guthabenzinsen werden im jeweiligen Jahr berücksichtigt |
| Abschreibung | 20.000 im ersten und jeweils 16.000 in den fünf weiteren Jahren: ausschließlich das **historische Buchschema** |
| Körperschaftsteuersatz | Die für 2027–2032 in § 23 KStG genannten Sätze |
| Zeitpunkt | Überschuss, Zinsen und die modellierte Steuerzahlung wirken jeweils zum Jahresende |

Die Beträge und Guthabenzinsen sind historische
Modellannahmen. Insbesondere prüfen wir mangels Angabe zur
Art der Investition **nicht**, ob das Abschreibungsschema für
eine tatsächliche Investition ab 2027 zulässig wäre.

Wir verwenden eine vereinfachte jährliche Ergebnisgröße:

**Investitionsüberschuss − Buchabschreibung
+ Guthabenzinsen − Kreditzinsen.**

Auf diese Größe wenden wir bedingt den jeweiligen
Körperschaftsteuersatz an. Wir setzen voraus, dass die
Kreditzinsen in diesem Fall abzugsfähig sind und dass
das im Buch ebenfalls beschriebene bestehende Geschäft
eine ausreichende positive Ergebnisbasis bietet.

Bei F5 werden die anfänglichen 100.000 eigenen Mittel
mitverzinst und die Guthabenzinsen in dieser Teilrechnung
mitbesteuert. Das ist für den Vergleich notwendig:
Die anfänglichen Mittel liegen bei F1 in der Investition,
bei F5 als Guthaben im Unternehmen.

Die Rechnung umfasst **keine vollständige Steuerlast**.
Insbesondere fehlen die Prüfung einer aktuellen
Abschreibung, Gewerbesteuer, Solidaritätszuschlag,
weitere steuerliche Anpassungen und eine mögliche
Besteuerung der Eigentümer bei Ausschüttung. Die
berechneten Endwerte heißen deshalb
*Endvermögen nach modellierter Körperschaftsteuer*.

Rechtsgrundlage für die verwendeten Sätze:
[§ 23 KStG](https://www.gesetze-im-internet.de/kstg_1977/__23.html).

```python
steuerplan_kapitalgesellschaft = (
    ergebnis_bruecke[
        ["jahr", "afa_buch"]
    ]
    .merge(
        koerperschaftsteuer_saetze[
            ["jahr", "kst_satz_prozent"]
        ],
        on="jahr",
    )
    .sort_values("jahr")
)

kst_simulation = []

for finanzierung, startguthaben, darlehen in [
    ("F1: eigene Mittel", 0.0, 0.0),
    ("F5: externer Kredit", eigene_mittel, kreditbetrag),
]:
    guthaben = startguthaben

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        guthaben_zu_jahresbeginn = guthaben
        guthabenzinsen = (
            guthaben_zu_jahresbeginn
            * guthabenzins_buch
        )
        kreditzinsen = darlehen * zins_fest_buch

        vereinfachtes_steuerergebnis = (
            ueberschuss_nominal
            - jahr.afa_buch
            + guthabenzinsen
            - kreditzinsen
        )

        modellierte_kst = (
            vereinfachtes_steuerergebnis
            * jahr.kst_satz_prozent
            / 100
        )

        tilgung = (
            darlehen if jahr.jahr == 2032 else 0.0
        )

        guthaben = (
            guthaben_zu_jahresbeginn
            + guthabenzinsen
            + ueberschuss_nominal
            - kreditzinsen
            - modellierte_kst
            - tilgung
        )

        kst_simulation.append({
            "finanzierung": finanzierung,
            "jahr": jahr.jahr,
            "guthaben_jahresbeginn": (
                guthaben_zu_jahresbeginn
            ),
            "guthabenzinsen": guthabenzinsen,
            "kreditzinsen": kreditzinsen,
            "afa_buch": jahr.afa_buch,
            "vereinfachtes_steuerergebnis": (
                vereinfachtes_steuerergebnis
            ),
            "modellierte_kst": modellierte_kst,
            "tilgung": tilgung,
            "endvermoegen_nach_modellierter_kst": guthaben,
        })

kst_simulation = pd.DataFrame(kst_simulation)

kst_simulation[
    [
        "finanzierung",
        "jahr",
        "guthabenzinsen",
        "kreditzinsen",
        "vereinfachtes_steuerergebnis",
        "modellierte_kst",
        "endvermoegen_nach_modellierter_kst",
    ]
].round(2)
```

### 11.1 Endvergleich der beiden Finanzierungen

Die jährliche Tabelle trennt Zinsen auf vorhandene Guthaben,
Kreditzinsen und die modellierte Körperschaftsteuer. Jetzt
vergleichen wir die Endbeträge des Jahres 2032. Für die
Kaufkraft verwenden wir erneut den Preisindex des
Inflations-Basispfads.

Dieser Vergleich gilt nur für die hier definierte
Kapitalgesellschaft mit im Unternehmen verbleibenden
Mitteln. Er beantwortet weder die Besteuerung bei
Ausschüttung noch den Vergleich mit Einzelunternehmen
und Personengesellschaft.

```python
kst_endvergleich = (
    kst_simulation.loc[
        kst_simulation["jahr"].eq(2032),
        [
            "finanzierung",
            "endvermoegen_nach_modellierter_kst",
        ],
    ]
    .copy()
)

kst_endvergleich[
    "kaufkraft_2026_im_basispfad"
] = (
    kst_endvergleich[
        "endvermoegen_nach_modellierter_kst"
    ]
    * 100
    / preisindex_basis_2032
)

vorteil_f1_nach_modellierter_kst = (
    kst_endvergleich.loc[
        kst_endvergleich["finanzierung"].eq(
            "F1: eigene Mittel"
        ),
        "endvermoegen_nach_modellierter_kst",
    ].iloc[0]
    -
    kst_endvergleich.loc[
        kst_endvergleich["finanzierung"].eq(
            "F5: externer Kredit"
        ),
        "endvermoegen_nach_modellierter_kst",
    ].iloc[0]
)

print(kst_endvergleich.round(2))
print(
    "Vorteil F1 nach modellierter Körperschaftsteuer:",
    round(vorteil_f1_nach_modellierter_kst, 2),
)
```

## 12. Gewerbesteuer im ersten Planjahr: Wirkung der Rechtsform

Die Körperschaftsteuerrechnung allein eignet sich nicht für
einen Vergleich zwischen Einzelunternehmen, Personengesellschaft
und Kapitalgesellschaft. Für einen angenommenen gewerblichen
Betrieb betrachten wir deshalb einen weiteren Steuerbaustein:
die Gewerbesteuer im Jahr 2027.

### Festlegungen für diese Teilrechnung

| Eingabe | Festlegung | Begründung |
|---|---|---|
| Tätigkeit | Gewerblicher Betrieb | Nur für diesen ausdrücklich angenommenen Fall rechnen wir hier Gewerbesteuer |
| Standort | Magdeburg | Konkreter fiktiver Beratungsstandort |
| Hebesatz 2027 | 450 % | Fortschreibung des von Magdeburg für **2026** veröffentlichten Werts; keine Festsetzung für 2027 |
| Einheit | Eine Geldeinheit wird in **dieser hypothetischen Steuerrechnung** als ein Euro behandelt | Steuerliche Freibeträge sind in Euro angegeben; die Gleichsetzung ist eine Modellentscheidung, keine Angabe des historischen Buchs |
| Geschäft vor der Investition | 247.000 jährlich | Referenzfall mit 13 % Rendite aus dem Buchbeispiel |
| Abschreibung | Historisches Buchschema | Dient nur der Illustration; aktuelle Zulässigkeit ist ungeprüft |
| Weitere Hinzurechnungen und Kürzungen | Keine angenommen | Über das bestehende Geschäft liegen uns hierzu keine Angaben vor |

Wir vergleichen F1 und F5 zunächst mit denselben
betrieblichen Ausgangszahlungen. Bei F5 bleiben die
100.000 eigenen Mittel im Betrieb angelegt. Deshalb
gehören ihre im Jahr 2027 modellierten Guthabenzinsen
zu dieser vereinfachten betrieblichen Ergebnisgröße.

Die jährlichen Kreditzinsen betragen 12.000. Für die
Teilrechnung unterstellen wir, dass keine weiteren
Aufwendungen vorliegen, die zusammen mit diesen Zinsen
eine gewerbesteuerliche Hinzurechnung nach § 8 GewStG
auslösen. Diese Annahme müsste für ein wirkliches
Unternehmen geprüft werden.

Wir runden den angenommenen Gewerbeertrag nach
§ 11 GewStG auf volle 100 Euro ab. Bei
Einzelunternehmen und Personengesellschaften
berücksichtigen wir dort den Freibetrag von 24.500 Euro;
bei der betrachteten Kapitalgesellschaft nicht.
Anschließend wenden wir die Steuermesszahl von 3,5 %
und den angenommenen Hebesatz von 450 % an.

Die Rechnung zeigt **nur die Gewerbesteuer dieses
hypothetischen Jahres**. Insbesondere ist eine mögliche
Ermäßigung der Einkommensteuer nach § 35 EStG noch
nicht berechnet. Aus einer niedrigeren Gewerbesteuer
allein folgt daher keine günstigere Rechtsform insgesamt.

Quellen:
[§ 8 GewStG](https://www.gesetze-im-internet.de/gewstg/__8.html),
[§ 11 GewStG](https://www.gesetze-im-internet.de/gewstg/__11.html),
[§ 16 GewStG](https://www.gesetze-im-internet.de/gewstg/__16.html),
[§ 35 EStG](https://www.gesetze-im-internet.de/estg/__35.html)
und [geänderte Haushaltssatzung Magdeburg 2026](
https://www.magdeburg.de/loadDocument.phtml?Ext=PDF&FID=37.51335.1).

```python
import math

hebesatz_magdeburg_modell = 4.50
steuermesszahl_gewerbe = 0.035
freibetrag_natuerliche_personen = 24_500

# Die bereits simulierten Ergebnisse vor modellierter KSt
# enthalten 2027 auch die Guthabenzinsen auf Betriebsvermögen.
basis_2027 = kst_simulation.loc[
    kst_simulation["jahr"].eq(2027),
    [
        "finanzierung",
        "vereinfachtes_steuerergebnis",
    ],
].copy()

basis_2027["vereinfachter_gewerbeertrag"] = (
    ueberschuss_alt_buch
    + basis_2027["vereinfachtes_steuerergebnis"]
)

gewerbesteuer_2027_zeilen = []

for zeile in basis_2027.itertuples(index=False):
    abgerundeter_ertrag = (
        math.floor(
            max(0, zeile.vereinfachter_gewerbeertrag)
            / 100
        )
        * 100
    )

    for rechtsform in [
        "Einzelunternehmen",
        "Personengesellschaft",
        "Kapitalgesellschaft",
    ]:
        freibetrag = (
            0
            if rechtsform == "Kapitalgesellschaft"
            else freibetrag_natuerliche_personen
        )

        bemessung_nach_freibetrag = max(
            0,
            abgerundeter_ertrag - freibetrag,
        )

        steuermessbetrag = (
            bemessung_nach_freibetrag
            * steuermesszahl_gewerbe
        )

        modellierte_gewerbesteuer = (
            steuermessbetrag
            * hebesatz_magdeburg_modell
        )

        gewerbesteuer_2027_zeilen.append({
            "rechtsform": rechtsform,
            "finanzierung": zeile.finanzierung,
            "vereinfachter_gewerbeertrag": (
                zeile.vereinfachter_gewerbeertrag
            ),
            "ertrag_abgerundet": abgerundeter_ertrag,
            "freibetrag": freibetrag,
            "steuermessbetrag": steuermessbetrag,
            "modellierte_gewerbesteuer": (
                modellierte_gewerbesteuer
            ),
        })

gewerbesteuer_2027 = pd.DataFrame(
    gewerbesteuer_2027_zeilen
)

gewerbesteuer_2027.round(2)
```

### 12.1 Interpretation der Gewerbesteuer-Teilrechnung

Für dieselbe vereinfachte betriebliche Ergebnisgröße
unterscheiden sich die berechneten Gewerbesteuerbeträge
zwischen den Rechtsformen wegen des Freibetrags.
Innerhalb einer Rechtsform verändert F5 den
angenommenen Gewerbeertrag gegenüber F1: Den
12.000 Kreditzinsen stehen im ersten Jahr
10.000 Guthabenzinsen auf die nicht eingesetzten
eigenen Mittel gegenüber.

Die Differenz zwischen den Rechtsformen ist **kein
Vergleich der gesamten Steuerbelastung**. Beim
Einzelunternehmen und bei der Personengesellschaft
fehlen die Einkommensteuer der Eigentümer und die
Prüfung der Ermäßigung nach § 35 EStG. Bei der
Kapitalgesellschaft fehlen unter anderem die
Besteuerung möglicher Ausschüttungen und weitere
steuerliche Anpassungen.

Außerdem stammen die Investitionsabschreibungen aus
einem historischen Buchbeispiel. Der Hebesatz für
2027 und die Behandlung der Geldeinheiten als Euro
sind ausdrücklich gesetzte Modellannahmen. Eine
Beratung für ein konkretes Unternehmen müsste
diese Eingaben ersetzen und den Rechtsstand
für das betreffende Jahr erneut prüfen.

## 13. Einkommensteuer der Eigentümer: fiktiver Haushalt im Jahr 2027

Um Einzelunternehmen und Personengesellschaft im selben
Beratungsfall zu betrachten, legen wir einen **fiktiven
Haushalt aus zwei zusammen veranlagten Ehepartnern** fest.

| Rechtsform | Eigentümer im Modell |
|---|---|
| Einzelunternehmen | Ein Ehepartner betreibt das Unternehmen; beide werden zusammen veranlagt |
| Personengesellschaft | Beide Ehepartner sind Mitunternehmer mit jeweils 50 % Gewinnanteil und werden zusammen veranlagt |

Der Haushalt besitzt in beiden Varianten dieselben
anfänglichen Mittel und erhält zusammen denselben
betrieblichen Gewinn. Die hälftige Beteiligung ist
eine einfache, ausdrücklich gesetzte Vergleichsannahme.
Wir untersuchen hier F1 und F5; ein
Gesellschafterdarlehen F4 benötigt zusätzlich die
Zuordnung von Zinsen und Darlehen zu einem der beiden
Mitunternehmer.

Für diese **hypothetische Rechnung** behandeln wir
den vereinfachten betrieblichen Gewinn aus Abschnitt 12
als einziges zu versteuerndes Einkommen des Haushalts.
Weitere Einkünfte, Abzüge und steuerliche Anpassungen
setzen wir nicht an. Dadurch ist die Rechnung
nachvollziehbar, bildet aber keine Steuererklärung ab.

Wir verwenden den bei Erstellung des Modells geltenden
Tarif nach § 32a EStG. Der Gesetzestext nennt die
Tarifformel **ab Veranlagungszeitraum 2026**. Ihre
Verwendung für unser Planjahr 2027 steht unter dem
Vorbehalt späterer Gesetzesänderungen. Nach dem
Splittingverfahren wird die Steuer für die Hälfte
des gemeinsamen zu versteuernden Einkommens ermittelt
und anschließend verdoppelt.

Danach zeigen wir eine **bedingte** Ermäßigung nach
§ 35 EStG. Da der Haushalt im Modell ausschließlich
die betrachteten gewerblichen Einkünfte hat,
setzen wir den Anteil dieser Einkünfte an seinen
positiven Einkünften zu 100 % an. Die Ermäßigung
begrenzen wir auf den kleinsten der folgenden Beträge:

1. das Vierfache des Gewerbesteuer-Messbetrags,
2. die modellierte tatsächlich zu zahlende Gewerbesteuer,
3. die tarifliche Einkommensteuer des Haushalts.

Schließlich addieren wir die verbleibende Einkommensteuer
und die Gewerbesteuer. Solidaritätszuschlag und
weitere mögliche Steuern sind nicht enthalten.
Die historischen Buchwerte und die angenommene
Abschreibung bleiben dieselben wie zuvor.

Rechtsgrundlagen:
[§ 26b EStG](https://www.gesetze-im-internet.de/estg/__26b.html),
[§ 32a EStG](https://www.gesetze-im-internet.de/estg/__32a.html)
und [§ 35 EStG](https://www.gesetze-im-internet.de/estg/__35.html).

```python
def est_grundtarif_tarifstand_2026(zve):
    """Tarifliche ESt nach § 32a Abs. 1 EStG,
    Tarifstand des Modells vom 24.09.2026.
    """
    x = math.floor(max(0, zve))

    if x <= 12_348:
        steuer = 0.0
    elif x <= 17_799:
        y = (x - 12_348) / 10_000
        steuer = (914.51 * y + 1_400) * y
    elif x <= 69_878:
        z = (x - 17_799) / 10_000
        steuer = (173.10 * z + 2_397) * z + 1_034.87
    elif x <= 277_825:
        steuer = 0.42 * x - 11_135.63
    else:
        steuer = 0.45 * x - 19_470.38

    return math.floor(steuer)


def est_splitting_tarifstand_2026(gemeinsames_zve):
    """Zweimal Grundtarif auf die Hälfte des
    gemeinsamen zu versteuernden Einkommens.
    """
    return 2 * est_grundtarif_tarifstand_2026(
        gemeinsames_zve / 2
    )
```

```python
haushaltssteuer_2027_zeilen = []

for fall in gewerbesteuer_2027.itertuples(index=False):
    if fall.rechtsform == "Kapitalgesellschaft":
        continue

    # Nur für dieses fiktive Profil:
    # Der vereinfachte betriebliche Gewinn ist das
    # gesamte zu versteuernde Einkommen des Haushalts.
    gemeinsames_zve = fall.vereinfachter_gewerbeertrag

    est_vor_ermaessigung = (
        est_splitting_tarifstand_2026(
            gemeinsames_zve
        )
    )

    ermaessigung_35 = min(
        4 * fall.steuermessbetrag,
        fall.modellierte_gewerbesteuer,
        est_vor_ermaessigung,
    )

    est_nach_ermaessigung = (
        est_vor_ermaessigung
        - ermaessigung_35
    )

    erfasste_steuern_zusammen = (
        est_nach_ermaessigung
        + fall.modellierte_gewerbesteuer
    )

    haushaltssteuer_2027_zeilen.append({
        "rechtsform": fall.rechtsform,
        "finanzierung": fall.finanzierung,
        "gemeinsames_zve_modell": gemeinsames_zve,
        "est_vor_paragraf_35": (
            est_vor_ermaessigung
        ),
        "ermaessigung_paragraf_35": (
            ermaessigung_35
        ),
        "est_nach_paragraf_35": (
            est_nach_ermaessigung
        ),
        "gewerbesteuer": (
            fall.modellierte_gewerbesteuer
        ),
        "erfasste_steuern_zusammen": (
            erfasste_steuern_zusammen
        ),
    })

haushaltssteuer_2027 = pd.DataFrame(
    haushaltssteuer_2027_zeilen
)

haushaltssteuer_2027.round(2)
```

### 13.1 Was dieser Vergleich zeigt

Unter dem gewählten Haushaltsprofil haben
Einzelunternehmen und Personengesellschaft bei F1
denselben gemeinsamen vereinfachten Gewinn. Dasselbe
gilt bei F5. Die gemeinsame Veranlagung und die
gleichen modellierten Gewerbesteuerbeträge führen
in dieser abgegrenzten Rechnung daher auch zu
gleichen erfassten Steuerbeträgen.

Das ist **kein allgemeiner steuerlicher Gleichstand
der Rechtsformen**. Andere Gewinnanteile,
unverheiratete Eigentümer, weitere Einkünfte,
Gesellschafterdarlehen oder unterschiedliche
Entnahmen können das Ergebnis verändern.

Wir dürfen diese Jahressteuer außerdem noch nicht
einfach vom Endvermögen des Jahres 2032 abziehen:
Dazu müssten die Steuerzahlungen und die dadurch
veränderten Guthabenzinsen in jedem Planjahr neu
berechnet werden. Die Kapitalgesellschaft braucht
zusätzlich ein einheitlich festgelegtes Vorgehen
für Gewinne, die im Unternehmen bleiben oder
an den Haushalt fließen.

## 14. Jährliche Steuern und Endvermögen des fiktiven Haushalts

Die bisherige Einkommensteuerrechnung betraf nur 2027.
Eine Steuerzahlung reduziert jedoch das Geld, das im
nächsten Jahr verzinst werden kann. Wir rechnen daher
F1 und den durchgehend mit 12 % verzinsten F5-Kredit
von 2027 bis 2032 jährlich weiter.

### Gleiche wirtschaftliche Ausgangslage

Die beiden zusammen veranlagten Ehepartner sind erneut
unsere gemeinsame wirtschaftliche Einheit. Beim
Einzelunternehmen betreibt eine Person das Geschäft;
bei der Personengesellschaft betreiben es beide mit
hälftigen Gewinnanteilen. In diesem abgegrenzten
F1/F5-Modell stehen dem Haushalt jeweils dieselben
100.000 eigenen Mittel zur Verfügung.

Das bestehende Geschäft liefert im Buch-Referenzfall
jedes Jahr 247.000 Geldeinheiten. Wir berechnen seine
modellierte Steuerlast auch **ohne** die neue Investition.
Anschließend berechnen wir die Steuerlast **mit**
Investition und Finanzierung. Die Differenz ist die
jährliche zusätzliche Steuerzahlung dieses Vorhabens.

Die Mittel aus dem bestehenden Geschäft und ihre
Steuerzahlungen behandeln wir als gemeinsamen
Hintergrund beider Alternativen; sie werden nicht
im Guthaben der neuen Investition angespart. So
addieren wir den alten Geschäftserfolg nicht nur
bei einer der beiden Finanzierungen zum Endvermögen.

Im Projektguthaben berücksichtigen wir jedes Jahr:

1. Guthabenzinsen auf den Stand zu Jahresbeginn,
2. den Investitionsüberschuss aus dem Buchbeispiel,
3. gegebenenfalls Bankzinsen,
4. die zusätzliche modellierte Steuerzahlung,
5. die Kredittilgung am Ende von 2032.

Wir verwenden dieselben vereinfachenden Steuerannahmen
wie in den Abschnitten 12 und 13: eine Geldeinheit
entspricht in dieser hypothetischen Rechnung einem Euro;
der Buch-Abschreibungsplan dient als Modellwert; der
Haushalt erzielt keine weiteren Einkünfte oder Abzüge.
Wir setzen keine weiteren gewerbesteuerlichen
Hinzurechnungen, Kürzungen oder Verlustvorträge an.

**Wichtige Fortschreibung:** Wir halten den derzeit
im Notebook verwendeten Einkommensteuertarif und
den als Modellwert übernommenen Hebesatz von 450 %
für alle sechs Planjahre fest. Das ist keine Aussage,
dass beide Größen künftig tatsächlich unverändert
bleiben. Eine Anwendung auf ein reales Unternehmen
erfordert für jedes Jahr den dann geltenden Rechtsstand.

Die Rechnung zeigt den **zusätzlichen Vermögenseffekt
der Investitions- und Finanzierungsentscheidung** für
den Haushalt unter diesen Annahmen. Das übrige
Haushaltsvermögen und Steuern außerhalb des Modells
sind nicht Teil des ausgewiesenen Endbetrags.

```python
def modellierte_haushaltssteuer(gewerblicher_gewinn):
    """ESt plus GewSt für das definierte fiktive
    Ehepaar mit ausschließlich gewerblichen Einkünften.
    Tarif und Hebesatz: konstante Modellannahmen.
    """
    abgerundeter_gewerbeertrag = (
        math.floor(
            max(0, gewerblicher_gewinn) / 100
        )
        * 100
    )

    steuermessbetrag = (
        max(
            0,
            abgerundeter_gewerbeertrag
            - freibetrag_natuerliche_personen
        )
        * steuermesszahl_gewerbe
    )

    gewerbesteuer = (
        steuermessbetrag
        * hebesatz_magdeburg_modell
    )

    est_vor_ermaessigung = (
        est_splitting_tarifstand_2026(
            gewerblicher_gewinn
        )
    )

    ermaessigung_35 = min(
        4 * steuermessbetrag,
        gewerbesteuer,
        est_vor_ermaessigung,
    )

    est_nach_ermaessigung = (
        est_vor_ermaessigung
        - ermaessigung_35
    )

    return {
        "gewerbesteuer": gewerbesteuer,
        "est_nach_paragraf_35": (
            est_nach_ermaessigung
        ),
        "steuern_zusammen": (
            gewerbesteuer
            + est_nach_ermaessigung
        ),
    }
```

```python
steuer_altgeschaeft = modellierte_haushaltssteuer(
    ueberschuss_alt_buch
)["steuern_zusammen"]

haushalt_verlauf_zeilen = []

for rechtsform in [
    "Einzelunternehmen",
    "Personengesellschaft",
]:
    for finanzierung, startguthaben, darlehen in [
        ("F1: eigene Mittel", 0.0, 0.0),
        (
            "F5: externer Kredit",
            eigene_mittel,
            kreditbetrag,
        ),
    ]:
        guthaben = startguthaben

        for jahr in steuerplan_kapitalgesellschaft.itertuples(
            index=False
        ):
            guthaben_anfang = guthaben
            guthabenzinsen = (
                guthaben_anfang
                * guthabenzins_buch
            )
            kreditzinsen = (
                darlehen * zins_fest_buch
            )

            # Vereinfachter Gewinn des gesamten
            # Betriebs einschließlich Altgeschäft.
            gewinn_mit_vorhaben = (
                ueberschuss_alt_buch
                + ueberschuss_nominal
                - jahr.afa_buch
                + guthabenzinsen
                - kreditzinsen
            )

            steuern_mit_vorhaben = (
                modellierte_haushaltssteuer(
                    gewinn_mit_vorhaben
                )["steuern_zusammen"]
            )

            zusaetzliche_steuern = (
                steuern_mit_vorhaben
                - steuer_altgeschaeft
            )

            tilgung = (
                darlehen
                if jahr.jahr == 2032
                else 0.0
            )

            guthaben = (
                guthaben_anfang
                + guthabenzinsen
                + ueberschuss_nominal
                - kreditzinsen
                - zusaetzliche_steuern
                - tilgung
            )

            haushalt_verlauf_zeilen.append({
                "rechtsform": rechtsform,
                "finanzierung": finanzierung,
                "jahr": jahr.jahr,
                "guthabenzinsen": guthabenzinsen,
                "kreditzinsen": kreditzinsen,
                "gewinn_mit_vorhaben": (
                    gewinn_mit_vorhaben
                ),
                "zusaetzliche_steuern": (
                    zusaetzliche_steuern
                ),
                "tilgung": tilgung,
                "projektvermoegen_jahresende": guthaben,
            })

haushalt_verlauf = pd.DataFrame(
    haushalt_verlauf_zeilen
)

haushalt_verlauf.round(2)
```

Wir zeigen für jede Kombination aus Rechtsform und

Der reale Wert verwendet erneut den Preisindex des
Inflations-Basispfads.

Wenn Einzelunternehmen und Personengesellschaft in

Beide Ehepartner werden gemeinsam veranlagt, der

Die ausgewiesenen Beträge beschreiben den zusätzlichen Vermögenseffekt

## 15. Entscheidung unter Zins- und Inflationsszenarien

Der bisherige Haushaltsvergleich setzt für F5 in jedem Jahr einen
Kreditzins von 12 % an. Für eine Beratung ist zusätzlich relevant,
welcher Kreditzins ab 2029 die Entscheidung verändern würde.

Wir halten den Kreditzins für **2027 und 2028 bei 12 %**, wie im
Buchreferenzfall. Für **2029 bis 2032** setzen wir probeweise 10 %,
12 % oder 14 % an. Diese Werte sind **Stressannahmen, keine
Zinsprognosen und kein konkretes Kreditangebot**. Sie zeigen, wie
empfindlich die Entscheidung auf einen um zwei Prozentpunkte
niedrigeren oder höheren späteren Zinssatz reagiert. Zusätzlich
berechnen wir den späteren Zins, bei dem F1 und F5 im vereinfachten
Steuermodell denselben Endbetrag erreichen.

In allen Fällen bleiben Investition, jährlicher Überschuss,
Guthabenzins, Abschreibungsannahme, Haushalt und steuerliche
Modellregeln gleich. Die bisher eingeführten Inflationspfade
verändern in dieser Rechnung ausschließlich die Kaufkraft der
Endbeträge: Wir unterstellen weiterhin **nominal feste
Projektüberschüsse und Zinssätze**. Deshalb folgt aus höherer
Inflation hier nicht automatisch ein höherer Kreditzins oder
ein höherer Umsatz.

Wir rechnen jede Variante erneut Jahr für Jahr, weil ein anderer
Kreditzins sowohl die Steuerzahlung als auch das anschließend
verzinste Guthaben beeinflusst.

```python
def endvermoegen_haushalt_f5(spaeterer_kreditzins):
    """F5 für den Haushalt: 12 % bis 2028,
    danach der als Argument übergebene Kreditzins.
    """
    guthaben = eigene_mittel

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        kreditzinssatz = (
            zins_fest_buch
            if jahr.jahr <= 2028
            else spaeterer_kreditzins
        )

        guthabenzinsen = guthaben * guthabenzins_buch
        kreditzinsen = kreditbetrag * kreditzinssatz

        gewinn_mit_vorhaben = (
            ueberschuss_alt_buch
            + ueberschuss_nominal
            - jahr.afa_buch
            + guthabenzinsen
            - kreditzinsen
        )

        zusaetzliche_steuern = (
            modellierte_haushaltssteuer(
                gewinn_mit_vorhaben
            )["steuern_zusammen"]
            - steuer_altgeschaeft
        )

        tilgung = (
            kreditbetrag if jahr.jahr == 2032 else 0.0
        )

        guthaben = (
            guthaben
            + guthabenzinsen
            + ueberschuss_nominal
            - kreditzinsen
            - zusaetzliche_steuern
            - tilgung
        )

    return guthaben


endvermoegen_f1_haushalt = (
    haushalt_endvergleich.loc[
        haushalt_endvergleich["rechtsform"].eq(
            "Einzelunternehmen"
        )
        & haushalt_endvergleich["finanzierung"].eq(
            "F1: eigene Mittel"
        ),
        "projektvermoegen_jahresende",
    ].iloc[0]
)

endvermoegen_f5_bisher = (
    haushalt_endvergleich.loc[
        haushalt_endvergleich["rechtsform"].eq(
            "Einzelunternehmen"
        )
        & haushalt_endvergleich["finanzierung"].eq(
            "F5: externer Kredit"
        ),
        "projektvermoegen_jahresende",
    ].iloc[0]
)

# Prüft, ob die neue Funktion den bisherigen
# Referenzfall mit 12 % in allen Jahren reproduziert.
assert abs(
    endvermoegen_haushalt_f5(zins_fest_buch)
    - endvermoegen_f5_bisher
) < 0.01

spaetere_zinsfaelle = {
    "10 % ab 2029": 0.10,
    "12 % ab 2029": 0.12,
    "14 % ab 2029": 0.14,
}

szenario_zeilen = []

for zinsfall, zinssatz in spaetere_zinsfaelle.items():
    endvermoegen_f5 = endvermoegen_haushalt_f5(
        zinssatz
    )

    for inflationsfall, preisindex in (
        preisindex_am_ende.items()
    ):
        faktor_kaufkraft = 100 / preisindex

        szenario_zeilen.append({
            "zinsfall": zinsfall,
            "inflationsfall": inflationsfall,
            "f1_nominal_2032": endvermoegen_f1_haushalt,
            "f5_nominal_2032": endvermoegen_f5,
            "vorteil_f1_nominal": (
                endvermoegen_f1_haushalt
                - endvermoegen_f5
            ),
            "f1_kaufkraft_2026": (
                endvermoegen_f1_haushalt
                * faktor_kaufkraft
            ),
            "f5_kaufkraft_2026": (
                endvermoegen_f5
                * faktor_kaufkraft
            ),
            "vorteil_f1_kaufkraft_2026": (
                (
                    endvermoegen_f1_haushalt
                    - endvermoegen_f5
                )
                * faktor_kaufkraft
            ),
        })

haushalt_szenarien = pd.DataFrame(
    szenario_zeilen
)

haushalt_szenarien.round(2)
```

```python
# Wir suchen den Zins mit gleichem Endvermögen
# durch schrittweises Halbieren eines Intervalls.
untere_grenze = 0.0
obere_grenze = 0.30

def abstand_f1_f5(spaeterer_zins):
    return (
        endvermoegen_f1_haushalt
        - endvermoegen_haushalt_f5(
            spaeterer_zins
        )
    )


if (
    abstand_f1_f5(untere_grenze)
    * abstand_f1_f5(obere_grenze)
    > 0
):
    raise ValueError(
        "Im untersuchten Zinsintervall liegt "
        "kein Gleichstand."
    )

for _ in range(60):
    mitte = (
        untere_grenze + obere_grenze
    ) / 2

    if (
        abstand_f1_f5(mitte)
        * abstand_f1_f5(untere_grenze)
        <= 0
    ):
        obere_grenze = mitte
    else:
        untere_grenze = mitte

zinsschwelle_nach_steuern = (
    untere_grenze + obere_grenze
) / 2

print(
    "Gleichstand bei Kreditzins ab 2029:",
    f"{100 * zinsschwelle_nach_steuern:.2f} %",
)

# Die Grafik verwendet dasselbe Steuermodell für
# jeden Zinspunkt; ihre Kurve ist keine Prognose.
zinsachse = [
    wert / 1000
    for wert in range(80, 141)
]

vorteile_f1 = [
    abstand_f1_f5(zinssatz)
    for zinssatz in zinsachse
]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    [100 * zinssatz for zinssatz in zinsachse],
    vorteile_f1,
    color="#20639B",
    linewidth=2.5,
    label="Vorteil F1 gegenüber F5",
)

ax.axhline(
    0,
    color="black",
    linewidth=1,
)

ax.axvline(
    100 * zinsschwelle_nach_steuern,
    color="#D1495B",
    linestyle="--",
    label=(
        "Gleichstand nach modellierten Steuern"
    ),
)

ax.axvline(
    12,
    color="gray",
    linestyle=":",
    label="12 %: Buchreferenz",
)

ax.set(
    title="Finanzierungsentscheidung des Haushalts",
    xlabel="Kreditzins 2029–2032 in %",
    ylabel="Endvermögen F1 minus F5 im Jahr 2032",
)

ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

### 15.1 Interpretation der Szenarien

Der modellierte Gleichstand zwischen F1 und F5 liegt bei einem
Kreditzins von **8,83 % für 2029 bis 2032**, wenn der Kreditzins
2027 und 2028 jeweils 12 % beträgt. Unterhalb dieser Schwelle
erreicht F5 im Modell ein höheres Endvermögen; oberhalb der
Schwelle erreicht F1 ein höheres Endvermögen.

Bei allen drei ausdrücklich untersuchten Zinssätzen von 10 %,
12 % und 14 % bleibt F1 vorteilhafter. Bei 10 % beträgt sein
nominaler Vorsprung **2.875,07**, bei 12 % **7.767,35** und bei
14 % **12.663,58** Geldeinheiten. Der Zinsfall von 10 % zeigt
also eine kleinere, aber noch keine umgekehrte Differenz.

Bei gleichem Kreditzins ist das nominale Endvermögen in allen
Inflationspfaden identisch. Höhere Inflation verringert lediglich
seine Kaufkraft in Geldeinheiten von 2026. Das ist eine Folge
unserer Annahme fester nominaler Überschüsse und Zinssätze.
Falls ein reales Unternehmen seine Preise, Kosten oder
Kreditkonditionen an die Inflation anpasst, müsste das Modell
diese Zusammenhänge ausdrücklich aufnehmen.

Die Zinsschwelle ist eine **bedingte Entscheidungshilfe** für
den definierten Haushalt und das vereinfachte Steuermodell.
Sie ist weder eine Zinsprognose noch ein allgemeiner
Grenzwert für Unternehmen.

## 16. Kapitalgesellschaft: Körperschaftsteuer und Gewerbesteuer

Abschnitt 11 hat für die Kapitalgesellschaft jährlich eine
vereinfachte Körperschaftsteuer berechnet. Nun ergänzen wir
für denselben Standort Magdeburg die Gewerbesteuer.

Wir verwenden wieder das bestehende Geschäft mit 247.000
Geldeinheiten Jahresüberschuss als gemeinsame Ausgangslage.
Für jedes Jahr berechnen wir die modellierte Körperschaft-
und Gewerbesteuer zunächst **ohne** das neue Vorhaben und
anschließend **mit** Vorhaben und Finanzierung. Nur die
Differenz belastet das ausgewiesene Projektguthaben.

| Festlegung | Begründung |
|---|---|
| Derselbe Betrieb, dieselbe Investition und dieselben F1/F5-Zahlungen | Die Finanzierungswege sollen innerhalb der Kapitalgesellschaft vergleichbar bleiben. |
| Körperschaftsteuersätze aus Abschnitt 4 | Wir führen die bereits dokumentierte gesetzliche Staffel im Modell fort. |
| Gewerbesteuer mit 3,5 % Messzahl und modelliertem Hebesatz von 450 % | Dieselben Eingaben wie in Abschnitt 12; der Hebesatz wird für die Planjahre lediglich fortgeschrieben. |
| Kein Freibetrag von 24.500 für die Kapitalgesellschaft | Dieser Freibetrag gilt in der hier betrachteten Regelung für natürliche Personen und Personengesellschaften. |
| Historischer Abschreibungsplan und keine weiteren gewerbesteuerlichen Anpassungen | Für den fiktiven Betrieb liegen keine genaueren Angaben vor. |
| Der zusätzliche Zahlungsmittelbestand bleibt bis Ende 2032 in der Gesellschaft | So lässt sich zunächst die Finanzierung **innerhalb** dieser Rechtsform untersuchen. |

Die Gewerbesteuer wird als Zahlung berücksichtigt, mindert
in dieser Rechnung aber nicht noch einmal die
Körperschaftsteuerbemessungsgrundlage. Ebenso wie im
bisherigen Haushaltsmodell lassen wir weitere Steuerbestandteile
und mögliche besondere Anpassungen offen.

**Vergleichsgrenze:** Das Ergebnis ist Geld *in der Gesellschaft*.
Die Endbeträge des Einzelunternehmens und der
Personengesellschaft aus Abschnitt 14 beschreiben dagegen
den modellierten Vermögenseffekt *für den Haushalt*.
Diese Beträge dürfen nicht unmittelbar als Rangfolge
der Rechtsformen gelesen werden. Dafür benötigen wir
eine festgelegte Ausschüttung und die zugehörige
Besteuerung der Eigentümer.

```python
def modellierte_gewerbesteuer_kapitalgesellschaft(
    gewerbeertrag
):
    """Vereinfachte GewSt der Kapitalgesellschaft
    mit den Annahmen aus Abschnitt 12.
    """
    ertrag_abgerundet = (
        math.floor(max(0, gewerbeertrag) / 100)
        * 100
    )

    return (
        ertrag_abgerundet
        * steuermesszahl_gewerbe
        * hebesatz_magdeburg_modell
    )


kapitalgesellschaft_zeilen = []

for finanzierung, startguthaben, darlehen in [
    ("F1: eigene Mittel", 0.0, 0.0),
    (
        "F5: externer Kredit",
        eigene_mittel,
        kreditbetrag,
    ),
]:
    guthaben = startguthaben

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        guthabenzinsen = (
            guthaben * guthabenzins_buch
        )
        kreditzinsen = (
            darlehen * zins_fest_buch
        )

        # Vereinfachte Bemessungsgrundlage des
        # gesamten Betriebs mit neuem Vorhaben.
        gewinn_mit_vorhaben = (
            ueberschuss_alt_buch
            + ueberschuss_nominal
            - jahr.afa_buch
            + guthabenzinsen
            - kreditzinsen
        )

        kst_satz = (
            jahr.kst_satz_prozent / 100
        )

        zusaetzliche_kst = (
            gewinn_mit_vorhaben
            - ueberschuss_alt_buch
        ) * kst_satz

        zusaetzliche_gewerbesteuer = (
            modellierte_gewerbesteuer_kapitalgesellschaft(
                gewinn_mit_vorhaben
            )
            - modellierte_gewerbesteuer_kapitalgesellschaft(
                ueberschuss_alt_buch
            )
        )

        tilgung = (
            darlehen if jahr.jahr == 2032 else 0.0
        )

        guthaben = (
            guthaben
            + guthabenzinsen
            + ueberschuss_nominal
            - kreditzinsen
            - zusaetzliche_kst
            - zusaetzliche_gewerbesteuer
            - tilgung
        )

        kapitalgesellschaft_zeilen.append({
            "finanzierung": finanzierung,
            "jahr": jahr.jahr,
            "guthabenzinsen": guthabenzinsen,
            "kreditzinsen": kreditzinsen,
            "zusaetzliche_kst": zusaetzliche_kst,
            "zusaetzliche_gewerbesteuer": (
                zusaetzliche_gewerbesteuer
            ),
            "tilgung": tilgung,
            "projektguthaben_in_gesellschaft": (
                guthaben
            ),
        })

kapitalgesellschaft_verlauf = pd.DataFrame(
    kapitalgesellschaft_zeilen
)

kapitalgesellschaft_verlauf.round(2)
```

```python
kapitalgesellschaft_endvergleich = (
    kapitalgesellschaft_verlauf.loc[
        kapitalgesellschaft_verlauf["jahr"].eq(2032),
        [
            "finanzierung",
            "projektguthaben_in_gesellschaft",
        ],
    ]
    .copy()
)

kapitalgesellschaft_endvergleich[
    "kaufkraft_2026_im_basispfad"
] = (
    kapitalgesellschaft_endvergleich[
        "projektguthaben_in_gesellschaft"
    ]
    * 100
    / preisindex_basis_2032
)

kapitalgesellschaft_endvergleich.round(2)
```

Nach der modellierten Körperschaft- und Gewerbesteuer
verbleiben Ende 2032 bei F1 **173.695,56** und bei F5
**163.394,10** Geldeinheiten im Projektguthaben der
Kapitalgesellschaft. F1 erreicht damit innerhalb dieser
Rechtsform einen um **10.301,46** Geldeinheiten höheren
Endbetrag.

In Abschnitt 11, der nur die Körperschaftsteuer
berücksichtigte, betrug der Vorsprung **13.082,81**.
Die ergänzte Gewerbesteuer verändert folglich nicht
die Rangfolge von F1 und F5 im Buchreferenzfall,
wohl aber die Höhe ihres Unterschieds.

Diese Beträge gehören am Ende des Planungszeitraums
weiterhin der Gesellschaft. Sie sind **keine
Auszahlungen an die Eigentümer**. Insbesondere wäre
es falsch, 173.695,56 unmittelbar mit dem
Haushaltsbetrag des Einzelunternehmens von 155.555,83
zu vergleichen und daraus eine Empfehlung für die
Rechtsform abzuleiten.

## 17. Verändert die Rechtsform die Finanzierungsgrenze?

Für den modellierten Haushalt haben wir einen
Gleichstand zwischen F1 und F5 bei **8,83 %
Kreditzins ab 2029** ermittelt. Für die
Kapitalgesellschaft berechnen wir dieselbe Frage
erneut. Der Kreditzins beträgt in beiden Rechnungen
2027 und 2028 jeweils 12 %; nur für 2029 bis 2032
wird er verändert.

Die beiden Rechnungen verwenden dieselbe Investition,
dieselben nominalen Überschüsse und denselben
Guthabenzins. Sie unterscheiden sich in den bislang
modellierten Steuern:

- Einzelunternehmen und Personengesellschaft:
  Einkommensteuer des festgelegten Haushalts,
  Gewerbesteuer und die modellierte Ermäßigung.
- Kapitalgesellschaft: Körperschaftsteuer und
  Gewerbesteuer; die Mittel bleiben in der Gesellschaft.

Eine unterschiedliche Zinsschwelle zeigt, **wie die
modellierte Steuerstruktur die Wahl zwischen F1 und F5
innerhalb der jeweiligen Rechtsform beeinflusst**.
Sie zeigt noch nicht, welche Rechtsform dem Haushalt
insgesamt den höchsten Auszahlungsbetrag verschafft.

```python
def endguthaben_kg_f5(spaeterer_kreditzins):
    """F5 innerhalb der Kapitalgesellschaft:
    12 % bis 2028, danach der übergebene Zinssatz.
    """
    guthaben = eigene_mittel

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        kreditzinssatz = (
            zins_fest_buch
            if jahr.jahr <= 2028
            else spaeterer_kreditzins
        )

        guthabenzinsen = (
            guthaben * guthabenzins_buch
        )
        kreditzinsen = (
            kreditbetrag * kreditzinssatz
        )

        gewinn_mit_vorhaben = (
            ueberschuss_alt_buch
            + ueberschuss_nominal
            - jahr.afa_buch
            + guthabenzinsen
            - kreditzinsen
        )

        zusaetzliche_kst = (
            gewinn_mit_vorhaben
            - ueberschuss_alt_buch
        ) * jahr.kst_satz_prozent / 100

        zusaetzliche_gewerbesteuer = (
            modellierte_gewerbesteuer_kapitalgesellschaft(
                gewinn_mit_vorhaben
            )
            - modellierte_gewerbesteuer_kapitalgesellschaft(
                ueberschuss_alt_buch
            )
        )

        tilgung = (
            kreditbetrag if jahr.jahr == 2032 else 0.0
        )

        guthaben = (
            guthaben
            + guthabenzinsen
            + ueberschuss_nominal
            - kreditzinsen
            - zusaetzliche_kst
            - zusaetzliche_gewerbesteuer
            - tilgung
        )

    return guthaben


endguthaben_kg_f1 = (
    kapitalgesellschaft_endvergleich.loc[
        kapitalgesellschaft_endvergleich[
            "finanzierung"
        ].eq("F1: eigene Mittel"),
        "projektguthaben_in_gesellschaft",
    ].iloc[0]
)

endguthaben_kg_f5_bisher = (
    kapitalgesellschaft_endvergleich.loc[
        kapitalgesellschaft_endvergleich[
            "finanzierung"
        ].eq("F5: externer Kredit"),
        "projektguthaben_in_gesellschaft",
    ].iloc[0]
)

# Derselbe Zinssatz in allen Jahren muss den
# bereits berechneten Referenzfall ergeben.
assert abs(
    endguthaben_kg_f5(zins_fest_buch)
    - endguthaben_kg_f5_bisher
) < 0.01


def abstand_kg_f1_f5(spaeterer_zins):
    return (
        endguthaben_kg_f1
        - endguthaben_kg_f5(spaeterer_zins)
    )


untere_kg = 0.0
obere_kg = 0.30

if (
    abstand_kg_f1_f5(untere_kg)
    * abstand_kg_f1_f5(obere_kg)
    > 0
):
    raise ValueError(
        "Im untersuchten Intervall liegt "
        "kein Gleichstand der Kapitalgesellschaft."
    )

for _ in range(60):
    mitte_kg = (
        untere_kg + obere_kg
    ) / 2

    if (
        abstand_kg_f1_f5(untere_kg)
        * abstand_kg_f1_f5(mitte_kg)
        <= 0
    ):
        obere_kg = mitte_kg
    else:
        untere_kg = mitte_kg

zinsschwelle_kg = (
    untere_kg + obere_kg
) / 2

vergleich_zinsschwellen = pd.DataFrame({
    "modell": [
        "Haushalt: Einzelunternehmen/Personengesellschaft",
        "Kapitalgesellschaft: Mittel verbleiben dort",
    ],
    "gleichstand_kreditzins_ab_2029_prozent": [
        100 * zinsschwelle_nach_steuern,
        100 * zinsschwelle_kg,
    ],
})

vergleich_zinsschwellen.round(2)
```

```python
zinsachse_prozent = [
    wert / 10
    for wert in range(80, 141)
]

abstand_haushalt = [
    abstand_f1_f5(zinssatz / 100)
    for zinssatz in zinsachse_prozent
]

abstand_gesellschaft = [
    abstand_kg_f1_f5(zinssatz / 100)
    for zinssatz in zinsachse_prozent
]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    zinsachse_prozent,
    abstand_haushalt,
    label="Haushalt",
    linewidth=2.5,
    color="#20639B",
)

ax.plot(
    zinsachse_prozent,
    abstand_gesellschaft,
    label="Kapitalgesellschaft: einbehaltene Mittel",
    linewidth=2.5,
    color="#D1495B",
)

ax.axhline(0, color="black", linewidth=1)
ax.axvline(
    12,
    color="gray",
    linestyle=":",
    label="12 %: Buchreferenz",
)

ax.set(
    title="Wann wechseln F1 und F5 ihre Rangfolge?",
    xlabel="Kreditzins 2029–2032 in %",
    ylabel="Endbetrag F1 minus F5",
)

ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

Der Gleichstand zwischen F1 und F5 liegt im modellierten

beträgt der Kreditzins in beiden Rechnungen weiterhin 12 %.

Die beiden Schwellen liegen in diesem konkreten Fall sehr
nah beieinander. Die Grafik zeigt zugleich, dass sich die

nach Steuerrechnung verändert. Aus der Nähe der

Für die Kapitalgesellschaft messen wir bisher Geld, das
Ende 2032 in der Gesellschaft verbleibt. Um dieselbe

## 18. Kapitalgesellschaft: Auszahlung an die Eigentümer

Bisher endete die Rechnung beim Guthaben der
Kapitalgesellschaft. Für den vereinbarten fiktiven
Fall ergänzen wir eine Auszahlung am Ende von 2032.

### Festgelegter Fall

| Frage | Modellannahme | Bedeutung |
|---|---|---|
| Wem gehört die Gesellschaft? | Den beiden Ehepartnern zu je 50 % | Wir betrachten denselben fiktiven Haushalt wie zuvor. |
| Woher stammen die 100.000 Anfangsmittel? | Aus bereits vorhandenen, einbehaltenen Gewinnen der Gesellschaft | Wir behandeln sie nicht als neue Einlage der Eigentümer. |
| Was geschieht bis 2032? | Das Projektguthaben bleibt in der Gesellschaft | Die bisherigen jährlichen Berechnungen bleiben die Grundlage. |
| Was geschieht Ende 2032? | Das gesamte ausgewiesene Projektguthaben wird nach der Kredittilgung ausgeschüttet | Erst jetzt rechnen wir eine Zahlung an den Haushalt. |
| Andere Kapitalerträge des Haushalts? | Keine | So steht der gemeinsame Sparer-Pauschbetrag in diesem Modell für die Ausschüttung zur Verfügung. |
| Steuerliche Behandlung der Auszahlung? | Vereinfachte Dividendenbesteuerung mit 25 % Einkommensteuer auf den Betrag nach Sparer-Pauschbetrag und 5,5 % Solidaritätszuschlag auf diese Steuer | Wir schreiben die heute verwendeten Regeln für den Planfall 2032 fort. |
| Weitere Wahlrechte und Abgaben? | Kein Antrag auf das Teileinkünfteverfahren; keine Kirchensteuer | Eine konkrete Beratung müsste diese Punkte anhand der Eigentümer prüfen. |

Da die Anfangsmittel nach unserer Annahme einbehaltene
Gewinne und keine neue Einlage sind, behandeln wir
das **gesamte ausgeschüttete Projektguthaben** in
dieser vereinfachten Rechnung als Dividende. Wir
unterstellen außerdem, dass die Auszahlung in dieser
Höhe gesellschaftsrechtlich möglich ist. Ein
Projektguthaben allein weist noch keinen tatsächlich
ausschüttungsfähigen Bilanzgewinn nach.

Die Eigentümersteuer fällt in diesem Modell erst bei
der Auszahlung Ende 2032 an. Wir ziehen sie deshalb
nicht rückwirkend von den Guthaben früherer Jahre ab.

**Grenze des Rechtsformvergleichs:** Die 100.000
befanden sich bei der Kapitalgesellschaft bereits
als einbehaltene Gewinne im Unternehmen. Im
Haushaltsmodell standen 100.000 eigene Mittel dem
Haushalt zur Verfügung. Das sind nicht automatisch
dieselben verfügbaren Anfangsvermögen der Eigentümer.
Die folgenden Nettoauszahlungen zeigen deshalb
zunächst den Vergleich **F1 gegen F5 innerhalb der
Kapitalgesellschaft**, noch keine faire Rangfolge
zwischen allen Rechtsformen.

```python
sparer_pauschbetrag_ehepaar = 2_000.00
steuersatz_kapitalertraege = 0.25
soli_satz_auf_kapitalertragsteuer = 0.055

auszahlung_zeilen = []

for fall in kapitalgesellschaft_endvergleich.itertuples(
    index=False
):
    brutto_auszahlung = (
        fall.projektguthaben_in_gesellschaft
    )

    steuerpflichtiger_betrag = max(
        0.0,
        brutto_auszahlung
        - sparer_pauschbetrag_ehepaar,
    )

    einkommensteuer_auf_dividende = (
        steuerpflichtiger_betrag
        * steuersatz_kapitalertraege
    )

    soli_auf_dividendensteuer = (
        einkommensteuer_auf_dividende
        * soli_satz_auf_kapitalertragsteuer
    )

    netto_auszahlung = (
        brutto_auszahlung
        - einkommensteuer_auf_dividende
        - soli_auf_dividendensteuer
    )

    auszahlung_zeilen.append({
        "finanzierung": fall.finanzierung,
        "brutto_auszahlung_2032": brutto_auszahlung,
        "sparer_pauschbetrag": (
            sparer_pauschbetrag_ehepaar
        ),
        "einkommensteuer_auf_dividende": (
            einkommensteuer_auf_dividende
        ),
        "soli_auf_dividendensteuer": (
            soli_auf_dividendensteuer
        ),
        "netto_an_haushalt_2032": netto_auszahlung,
        "kaufkraft_netto_2026": (
            netto_auszahlung
            * 100
            / preisindex_basis_2032
        ),
    })

ausschuettung_2032 = pd.DataFrame(
    auszahlung_zeilen
)

ausschuettung_2032.round(2)
```

### 18.1 Interpretation der Auszahlung

Nach der modellierten Ausschüttung Ende 2032 erhält
der gemeinsame Haushalt bei F1 **128.410,86** und bei
F5 **120.826,40** Geldeinheiten. Innerhalb der
Kapitalgesellschaft führt F1 somit zu einer um
**7.584,46** Geldeinheiten höheren Nettoauszahlung.

Vor der Ausschüttung betrug der Unterschied zwischen
F1 und F5 **10.301,46** Geldeinheiten. Die
modellierte Steuer auf die Auszahlung verringert
diesen Unterschied, kehrt die Rangfolge aber nicht um.

Ein direkter Vergleich der Nettoauszahlung mit den
Endbeträgen des Einzelunternehmens oder der
Personengesellschaft wäre noch keine faire
Rechtsformentscheidung. In der Kapitalgesellschaft
lagen die anfänglichen 100.000 bereits als
einbehaltene Gewinne vor; im anderen Haushaltsmodell
waren 100.000 eigene Mittel verfügbar. Gleiche
nominale Anfangsbeträge bedeuten hier nicht
automatisch gleiches verfügbares Vermögen der
Eigentümer im Jahr 2026.

Außerdem berücksichtigt die vereinfachte Rechnung
nicht alle möglichen Steuern und Wahlrechte.
Insbesondere haben wir den Solidaritätszuschlag auf
die jährliche Körperschaftsteuer nicht ergänzt und
keinen Antrag auf eine andere Besteuerung der
Ausschüttung untersucht. Eine Empfehlung für die
Rechtsform lässt sich aus den ausgewiesenen
Auszahlungsbeträgen daher nicht ableiten.

## 19. Bedingte Beratungsentscheidung für F1 und F5

Wir fassen die untersuchten späteren Kreditzinsen
zusammen. Für jede Rechtsformrechnung vergleichen
wir F1 und F5 **innerhalb desselben Modells**:

1. Beim Einzelunternehmen und bei der
   Personengesellschaft betrachten wir das
   zusätzliche Endvermögen des fiktiven Haushalts.
2. Bei der Kapitalgesellschaft betrachten wir die
   Nettoauszahlung an denselben Haushalt nach der
   modellierten Ausschüttung Ende 2032.

Der Kreditzins beträgt 2027 und 2028 jeweils 12 %.
Für 2029 bis 2032 betrachten wir 10 %, 12 % und
14 % als Stressfälle, nicht als Prognosen. Die
Tabelle nennt für jeden Fall die innerhalb des
jeweiligen Modells günstigere Finanzierung.

**Leseregel:** Eine Zeile darf F1 mit F5 vergleichen.
Die Endbeträge unterschiedlicher Rechtsformen
dürfen hier noch nicht zur Wahl einer Rechtsform
gegeneinander aufgerechnet werden.

```python
def netto_auszahlung_kg(brutto_auszahlung):
    """Eigentümerrechnung aus Abschnitt 18
    für eine vollständige Auszahlung 2032.
    """
    steuerpflichtig = max(
        0.0,
        brutto_auszahlung
        - sparer_pauschbetrag_ehepaar,
    )

    kapitalertragsteuer_modell = (
        steuerpflichtig
        * steuersatz_kapitalertraege
    )

    soli_modell = (
        kapitalertragsteuer_modell
        * soli_satz_auf_kapitalertragsteuer
    )

    return (
        brutto_auszahlung
        - kapitalertragsteuer_modell
        - soli_modell
    )


f1_kg_netto = (
    ausschuettung_2032.loc[
        ausschuettung_2032["finanzierung"].eq(
            "F1: eigene Mittel"
        ),
        "netto_an_haushalt_2032",
    ].iloc[0]
)

f5_kg_netto_bisher = (
    ausschuettung_2032.loc[
        ausschuettung_2032["finanzierung"].eq(
            "F5: externer Kredit"
        ),
        "netto_an_haushalt_2032",
    ].iloc[0]
)

assert abs(
    netto_auszahlung_kg(
        endguthaben_kg_f5(zins_fest_buch)
    )
    - f5_kg_netto_bisher
) < 0.01

entscheidung_zeilen = []

for zinsfall, spaeterer_zins in (
    spaetere_zinsfaelle.items()
):
    faelle = [
        {
            "modell": (
                "Einzelunternehmen / "
                "Personengesellschaft"
            ),
            "f1": endvermoegen_f1_haushalt,
            "f5": endvermoegen_haushalt_f5(
                spaeterer_zins
            ),
        },
        {
            "modell": (
                "Kapitalgesellschaft: Auszahlung 2032"
            ),
            "f1": f1_kg_netto,
            "f5": netto_auszahlung_kg(
                endguthaben_kg_f5(
                    spaeterer_zins
                )
            ),
        },
    ]

    for fall in faelle:
        vorteil_f1 = (
            fall["f1"] - fall["f5"]
        )

        entscheidung_zeilen.append({
            "modell": fall["modell"],
            "kreditzinsfall": zinsfall,
            "f1_endbetrag": fall["f1"],
            "f5_endbetrag": fall["f5"],
            "vorteil_f1": vorteil_f1,
            "guenstiger_innerhalb_des_modells": (
                "F1" if vorteil_f1 > 0
                else "F5" if vorteil_f1 < 0
                else "Gleichstand"
            ),
        })

entscheidung_f1_f5 = pd.DataFrame(
    entscheidung_zeilen
)

entscheidung_f1_f5.round(2)
```

### 19.1 Beratungsfazit für die untersuchten Finanzierungen

Für die Investition aus dem historischen Buchbeispiel
ist F1 unter allen drei untersuchten späteren
Kreditzinsen von 10 %, 12 % und 14 % günstiger als F5.
Das gilt sowohl für den festgelegten Haushalt mit
Einzelunternehmen oder Personengesellschaft als auch
für die Kapitalgesellschaft mit vollständiger
Ausschüttung Ende 2032.

Der Vorteil ist **bedingt**: Bis einschließlich 2028
beträgt der Kreditzins 12 %. Erst der Zins von 2029
bis 2032 verändert sich. Im Haushaltsmodell wechseln
F1 und F5 bei ungefähr 8,83 % ihre Rangfolge;
innerhalb der Kapitalgesellschaft liegt die
berechnete Grenze bei ungefähr 8,82 %. Die
untersuchten 10 %, 12 % und 14 % liegen darüber.

Die Inflationspfade verändern bei unseren festen
nominalen Zahlungen die Kaufkraft, aber nicht die
Rangfolge. Wir haben keine Eintrittswahrscheinlichkeiten
für die Szenarien geschätzt und keine eigene
Inflations- oder Zinsprognose erstellt. Die Analyse
ist eine **deterministische Szenario- und
Sensitivitätsrechnung**, keine Zufallssimulation.

Aus der Tabelle folgt **keine Empfehlung für eine
Rechtsform**. Die vorhandenen 100.000 liegen im
Kapitalgesellschaftsfall als einbehaltene Gewinne
bereits im Unternehmen. Das Haushaltsmodell setzt
100.000 verfügbare eigene Mittel voraus. Zudem
bildet das Steuerteilmodell nicht sämtliche
Steuerzahlungen und Wahlrechte ab. Beträge aus
verschiedenen Rechtsformrechnungen sind daher
nicht als gleich ausgestattete Alternativen zu lesen.

## 20. Beratungsfrage: Welche Mittel sind tatsächlich verfügbar?

Bevor F1, F2, F4 und F5 gemeinsam verglichen werden,
fragt der Berater:

> „Das Unternehmen verfügt bereits über 100.000
> Geldeinheiten. Verfügen Sie als Eigentümer **zusätzlich**
> über 100.000 Geldeinheiten privat, die Sie für die
> Finanzierung einsetzen könnten?“

Wir untersuchen beide möglichen Antworten:

| Antwort | Mittel im Unternehmen | Zusätzliche private Mittel | Finanzierungen, die aus diesen Mitteln zunächst möglich sind |
|---|---:|---:|---|
| Nein | 100.000 | 0 | F1 oder F5 |
| Ja | 100.000 | 100.000 | F1, F2, F4 oder F5 |

**F1:** Das Unternehmen verwendet seine vorhandenen Mittel.
**F2:** Die Eigentümer bringen ihre zusätzlichen privaten
Mittel als Eigenkapital ein. **F4:** Sie geben der
Gesellschaft stattdessen ein Darlehen. **F5:** Die Bank
finanziert die Investition; vorhandene Mittel bleiben
zunächst verfügbar.

Die Tabelle prüft zunächst nur die **Finanzierbarkeit
zum Entscheidungszeitpunkt Ende 2026**. Bei F4 sind
die Darlehensforderung des Haushalts und die
entsprechende Schuld des Unternehmens gemeinsam
betrachtet kein zusätzliches Vermögen. Bei F5 ist
die Bankschuld dagegen eine Schuld gegenüber einem
Dritten.

Wir bewerten die neue Investition unmittelbar nach
der Finanzierung vereinfachend mit ihren
Anschaffungskosten von 100.000. Damit prüfen wir,
ob die Alternativen eines jeweiligen Falls mit
demselben Anfangsvermögen starten. Jahreserträge,
Zinsen und Steuern folgen erst im nächsten Schritt.
Für F2 und F4 müssen Einlage, Darlehen, Rückzahlung
und Eigentümersteuer getrennt erfasst werden;
insbesondere darf eine spätere Rückzahlung nicht
pauschal als Gewinnausschüttung behandelt werden.

**Wichtig für die bisherigen Ergebnisse:** Die
F1/F5-Rechnungen beziehen sich bislang auf den
Fall ohne zusätzliche private 100.000. Falls
zusätzliche private Mittel vorhanden sind, müssen
wir sie auch bei F1 und F5 berücksichtigen, selbst
wenn sie dort nicht zur Investition verwendet
werden. Die bisherigen Endbeträge dürfen deshalb
nicht unverändert neben neue F2/F4-Endbeträge
gestellt werden.

```python
investitionsbetrag = investition
mittel_im_unternehmen_2026 = eigene_mittel

private_faelle = {
    "Nein: keine zusätzlichen Privatmittel": 0.0,
    "Ja: zusätzliche 100.000 privat": eigene_mittel,
}

anfangsbilanz_zeilen = []

for antwort, private_mittel in private_faelle.items():
    for weg in ["F1", "F2", "F4", "F5"]:
        braucht_private_mittel = weg in ["F2", "F4"]

        finanzierbar = (
            not braucht_private_mittel
            or private_mittel >= investitionsbetrag
        )

        if not finanzierbar:
            anfangsbilanz_zeilen.append({
                "antwort": antwort,
                "finanzierung": weg,
                "status": (
                    "Nicht aus vorhandenen "
                    "Privatmitteln finanzierbar"
                ),
                "freie_mittel_unternehmen": None,
                "freie_mittel_privat": None,
                "investition_zu_kosten": None,
                "bankverbindlichkeit": None,
                "forderung_eigentuemer": None,
                "schuld_an_eigentuemer": None,
                "gemeinsames_nettovermoegen": None,
            })
            continue

        freie_mittel_unternehmen = (
            mittel_im_unternehmen_2026
        )
        freie_mittel_privat = private_mittel

        bankverbindlichkeit = 0.0
        forderung_eigentuemer = 0.0
        schuld_an_eigentuemer = 0.0

        if weg == "F1":
            freie_mittel_unternehmen -= (
                investitionsbetrag
            )

        elif weg == "F2":
            freie_mittel_privat -= (
                investitionsbetrag
            )
            # Private Einlage kommt zunächst an;
            # anschließend bezahlt das Unternehmen
            # damit die Investition. Sein zuvor
            # vorhandenes Geld bleibt verfügbar.

        elif weg == "F4":
            freie_mittel_privat -= (
                investitionsbetrag
            )
            forderung_eigentuemer = (
                investitionsbetrag
            )
            schuld_an_eigentuemer = (
                investitionsbetrag
            )

        elif weg == "F5":
            bankverbindlichkeit = (
                investitionsbetrag
            )

        gemeinsames_nettovermoegen = (
            freie_mittel_unternehmen
            + freie_mittel_privat
            + investitionsbetrag
            + forderung_eigentuemer
            - schuld_an_eigentuemer
            - bankverbindlichkeit
        )

        anfangsbilanz_zeilen.append({
            "antwort": antwort,
            "finanzierung": weg,
            "status": "Finanzierbar",
            "freie_mittel_unternehmen": (
                freie_mittel_unternehmen
            ),
            "freie_mittel_privat": (
                freie_mittel_privat
            ),
            "investition_zu_kosten": (
                investitionsbetrag
            ),
            "bankverbindlichkeit": (
                bankverbindlichkeit
            ),
            "forderung_eigentuemer": (
                forderung_eigentuemer
            ),
            "schuld_an_eigentuemer": (
                schuld_an_eigentuemer
            ),
            "gemeinsames_nettovermoegen": (
                gemeinsames_nettovermoegen
            ),
        })

anfangsbilanz = pd.DataFrame(
    anfangsbilanz_zeilen
)

anfangsbilanz.round(2)
```

Die Tabelle beantwortet zunächst die Frage nach der

- **Ohne zusätzliche private Mittel** sind F1 und F5
  möglich. Beide starten mit 100.000 Geldeinheiten
  gemeinsamem Nettovermögen.
- **Mit zusätzlichen 100.000 privaten Mitteln** sind
  F1, F2, F4 und F5 möglich. Alle vier starten mit

Die gleiche Anfangssumme genügt noch nicht für einen
Endvergleich: Nach der Finanzierung liegt das freie
Geld je nach Weg im Unternehmen oder beim Haushalt.
Die Erträge und ihre steuerliche Behandlung können
sich deshalb unterscheiden.

Wir führen zuerst F1 und F5 für **beide Antworten**
bis 2032 fort. Im Ja-Fall bleibt bei diesen Wegen
der zusätzliche private Betrag außerhalb des
Unternehmens angelegt. F2 und F4 benötigen
anschließend eigene Zahlungswege für Einlage
beziehungsweise Gesellschafterdarlehen.

## 21. F1 und F5 mit und ohne zusätzliche private Mittel

Für den Nein-Fall verwenden wir die bereits
berechneten betrieblichen Endbeträge. Im Ja-Fall
kommt bei F1 und F5 ein privates Anfangsguthaben
von 100.000 hinzu. Es wird im Modell mit dem
Guthabenzins von 10 % aus dem historischen
Buchbeispiel angelegt.

Wir berechnen die privaten Guthabenzinsen jährlich.
Für den fiktiven Haushalt nehmen wir weiterhin
keine anderen privaten Kapitalerträge an. Auf die
Kapitalerträge eines Jahres wenden wir den
gemeinsamen Sparer-Pauschbetrag von 2.000 und
anschließend die im Modell verwendete
Kapitalertragsteuer und den Solidaritätszuschlag an.

Bei der Kapitalgesellschaft fließt die modellierte
Ausschüttung Ende 2032 hinzu. **In diesem Jahr
werden private Guthabenzinsen und Ausschüttung
zusammen betrachtet**, damit der Sparer-Pauschbetrag
nicht zweimal genutzt wird. Für die früheren Jahre
gibt es in diesem Teilmodell keine Ausschüttung.

Die Beträge aus unterschiedlichen Rechtsformmodellen
bleiben trotz dieser Ergänzung nicht automatisch
vergleichbar: Die 100.000 im Unternehmen stammen
im Kapitalgesellschaftsfall aus bereits
einbehaltenen Gewinnen, und weitere Steuerbestandteile
sind offen. Innerhalb eines Falls und einer
Rechtsformrechnung können wir aber erneut F1 und F5
vergleichen.

```python
def privater_endbetrag(
    anfangsguthaben,
    ausschuettung_2032_brutto=0.0,
):
    """Privates Konto 2027–2032.

    2032 teilen sich private Guthabenzinsen und
    Ausschüttung denselben Sparer-Pauschbetrag.
    """
    guthaben = anfangsguthaben

    for jahr in range(2027, 2033):
        private_zinsen = (
            guthaben * guthabenzins_buch
        )

        ausschuettung = (
            ausschuettung_2032_brutto
            if jahr == 2032
            else 0.0
        )

        kapitalertraege = (
            private_zinsen + ausschuettung
        )

        steuerpflichtig = max(
            0.0,
            kapitalertraege
            - sparer_pauschbetrag_ehepaar,
        )

        kapitalertragsteuer = (
            steuerpflichtig
            * steuersatz_kapitalertraege
        )

        soli = (
            kapitalertragsteuer
            * soli_satz_auf_kapitalertragsteuer
        )

        guthaben = (
            guthaben
            + private_zinsen
            + ausschuettung
            - kapitalertragsteuer
            - soli
        )

    return guthaben


# Ohne privates Anfangsguthaben muss die Funktion
# die frühere Nettoauszahlung wiedergeben.
assert abs(
    privater_endbetrag(
        0.0,
        endguthaben_kg_f1,
    )
    - f1_kg_netto
) < 0.01

assert abs(
    privater_endbetrag(
        0.0,
        endguthaben_kg_f5_bisher,
    )
    - f5_kg_netto_bisher
) < 0.01


vergleich_antworten_zeilen = []

for antwort, private_mittel in (
    private_faelle.items()
):
    faelle_f1_f5 = [
        {
            "modell": (
                "Einzelunternehmen / "
                "Personengesellschaft"
            ),
            "finanzierung": "F1",
            "endbetrag": (
                endvermoegen_f1_haushalt
                + privater_endbetrag(
                    private_mittel
                )
            ),
        },
        {
            "modell": (
                "Einzelunternehmen / "
                "Personengesellschaft"
            ),
            "finanzierung": "F5",
            "endbetrag": (
                endvermoegen_f5_bisher
                + privater_endbetrag(
                    private_mittel
                )
            ),
        },
        {
            "modell": (
                "Kapitalgesellschaft: "
                "Ausschüttung 2032"
            ),
            "finanzierung": "F1",
            "endbetrag": (
                privater_endbetrag(
                    private_mittel,
                    endguthaben_kg_f1,
                )
            ),
        },
        {
            "modell": (
                "Kapitalgesellschaft: "
                "Ausschüttung 2032"
            ),
            "finanzierung": "F5",
            "endbetrag": (
                privater_endbetrag(
                    private_mittel,
                    endguthaben_kg_f5_bisher,
                )
            ),
        },
    ]

    for fall in faelle_f1_f5:
        vergleich_antworten_zeilen.append({
            "antwort": antwort,
            "modell": fall["modell"],
            "finanzierung": fall["finanzierung"],
            "endbetrag_2032": fall["endbetrag"],
            "kaufkraft_2026_basis": (
                fall["endbetrag"]
                * 100
                / preisindex_basis_2032
            ),
        })

vergleich_antworten = pd.DataFrame(
    vergleich_antworten_zeilen
)

vergleich_antworten.round(2)
```

## 22. F2 im Ja-Fall: zusätzliche Einlage der Eigentümer

Nun nutzen die Eigentümer die zusätzlich
vorhandenen **privaten 100.000**, um die
Investition zu finanzieren. Das Unternehmen
besitzt außerdem schon 100.000. Diese bleiben
im Modell auf einem betrieblichen Guthaben
angelegt.

Für den fiktiven Haushalt setzen wir fest:

| Punkt | Annahme |
|---|---|
| Einzelunternehmen | Der betreibende Ehepartner führt 100.000 aus privaten Mitteln zu. |
| Personengesellschaft | Beide Ehepartner führen jeweils 50.000 zu; ihre bisherigen Gewinnanteile bleiben bei je 50 %. |
| Auszahlung der Einlage | Keine Entnahme während 2027–2032; das gemeinsame betriebliche Geld zählt zum Endvermögen des Haushalts. |
| Anlage des bisherigen Unternehmensguthabens | 10 % jährlich wie im historischen Buchbeispiel; kein aktuelles Anlageangebot. |
| Projekt und Steuern | Dieselben Überschüsse, Abschreibungsannahmen und vereinfachten Haushaltssteuern wie bei F1/F5. |

Die Einlage selbst behandeln wir nicht als
steuerpflichtigen Projektertrag. Die
Guthabenzinsen des im Betrieb verbliebenen
Geldes gehen dagegen in die bisher modellierte
betriebliche Ergebnisgröße ein.

F2 wird **nur für die Antwort „Ja“** berechnet.
Im Nein-Fall fehlen die zusätzlichen privaten
100.000. Für die Kapitalgesellschaft folgt
später eine eigene Rechnung: Bei einer
Ausschüttung darf eine neue Einlage nicht
automatisch genauso behandelt werden wie
einbehaltene Gewinne.

```python
f2_haushalt_zeilen = []

for rechtsform in [
    "Einzelunternehmen",
    "Personengesellschaft",
]:
    # Die privaten 100.000 bezahlen die Investition.
    # Das bereits vorhandene Unternehmensguthaben
    # von 100.000 bleibt im Betrieb.
    betriebsguthaben = (
        mittel_im_unternehmen_2026
    )

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        guthabenzinsen_betrieb = (
            betriebsguthaben
            * guthabenzins_buch
        )

        gewinn_mit_vorhaben = (
            ueberschuss_alt_buch
            + ueberschuss_nominal
            - jahr.afa_buch
            + guthabenzinsen_betrieb
        )

        zusaetzliche_steuern = (
            modellierte_haushaltssteuer(
                gewinn_mit_vorhaben
            )["steuern_zusammen"]
            - steuer_altgeschaeft
        )

        betriebsguthaben = (
            betriebsguthaben
            + guthabenzinsen_betrieb
            + ueberschuss_nominal
            - zusaetzliche_steuern
        )

        f2_haushalt_zeilen.append({
            "rechtsform": rechtsform,
            "jahr": jahr.jahr,
            "guthabenzinsen_betrieb": (
                guthabenzinsen_betrieb
            ),
            "gewinn_mit_vorhaben": (
                gewinn_mit_vorhaben
            ),
            "zusaetzliche_steuern": (
                zusaetzliche_steuern
            ),
            "endvermoegen_haushalt": (
                betriebsguthaben
            ),
        })

f2_haushalt_verlauf = pd.DataFrame(
    f2_haushalt_zeilen
)

f2_haushalt_verlauf.round(2)
```

```python
f2_endvergleich_haushalt = (
    f2_haushalt_verlauf.loc[
        f2_haushalt_verlauf["jahr"].eq(2032),
        [
            "rechtsform",
            "endvermoegen_haushalt",
        ],
    ]
    .copy()
)

f2_endvergleich_haushalt[
    "kaufkraft_2026_basis"
] = (
    f2_endvergleich_haushalt[
        "endvermoegen_haushalt"
    ]
    * 100
    / preisindex_basis_2032
)

ja_haushalt_f1_f5 = (
    vergleich_antworten.loc[
        vergleich_antworten["antwort"].eq(
            "Ja: zusätzliche 100.000 privat"
        )
        & vergleich_antworten["modell"].eq(
            "Einzelunternehmen / "
            "Personengesellschaft"
        ),
        [
            "finanzierung",
            "endbetrag_2032",
        ],
    ]
    .copy()
)

f2_betrag = (
    f2_endvergleich_haushalt.loc[
        f2_endvergleich_haushalt[
            "rechtsform"
        ].eq("Einzelunternehmen"),
        "endvermoegen_haushalt",
    ].iloc[0]
)

ja_haushalt_vergleich = pd.concat([
    ja_haushalt_f1_f5,
    pd.DataFrame([{
        "finanzierung": "F2",
        "endbetrag_2032": f2_betrag,
    }]),
], ignore_index=True)

ja_haushalt_vergleich[
    "kaufkraft_2026_basis"
] = (
    ja_haushalt_vergleich[
        "endbetrag_2032"
    ]
    * 100
    / preisindex_basis_2032
)

print(f2_endvergleich_haushalt.round(2))
print(ja_haushalt_vergleich.round(2))
```

## 23. F4 im Ja-Fall: Darlehen der Gesellschafter

> **Frage des Beraters:** „Wenn Sie privat zusätzlich 100.000
> besitzen: Möchten Sie diese der Personengesellschaft als
> Einlage überlassen oder ihr stattdessen ein Darlehen geben?“

Für F4 geben die beiden Ehepartner der Personengesellschaft
Ende 2026 jeweils 50.000 als Darlehen. Die Gesellschaft bezahlt
damit die Investition. Ihre bereits vorhandenen 100.000 bleiben
als betriebliches Guthaben angelegt.

| Punkt | Modellannahme und Begründung |
|---|---|
| Darlehensbetrag | Zusammen 100.000; genau der Betrag der Investition. |
| Darlehenszins | Fest 12 % pro Jahr als Vergleichswert aus dem historischen Buchbeispiel, nicht als aktueller Markt- oder Fremdvergleichszins. |
| Tilgung | Der gesamte Darlehensbetrag wird Ende 2032 zurückgezahlt; so stimmt der Zeithorizont mit den bisherigen Rechnungen überein. |
| Jährliche Zinsen | Die Gesellschaft zahlt 12.000 an die Gesellschafter. Diese legen erhaltenes Geld ab dem Folgejahr privat zu den angenommenen 10 % Guthabenzins an. |
| Steuerliche Zuordnung | Darlehenszinsen an Mitunternehmer behandeln wir als gewerbliche Sondervergütungen. Der Zinsaufwand der Gesellschaft und die Sondervergütung heben sich in unserer zusammengefassten Gewinnrechnung des Haushalts auf. |
| Private Kapitalerträge | Nur die Zinsen auf dem privaten Bankguthaben durchlaufen die bisher verwendete vereinfachte Kapitalertragsteuerrechnung. |
| Endvermögen | Betriebliches Guthaben plus privates Guthaben nach Rückzahlung des Darlehens. Die Darlehensforderung wird nicht zusätzlich addiert. |

Die jährlichen 12.000 sind eine Zahlung **innerhalb des betrachteten
Haushalts und Unternehmens**. Sie dürfen dessen gemeinsames
Vermögen nicht allein durch die Überweisung erhöhen. Sie verändern
aber, wo das Geld anschließend liegt und wie spätere Bankzinsen
im Modell besteuert werden.

Rechtsgrundlagen für die steuerliche Einordnung:
[§ 15 EStG](https://www.gesetze-im-internet.de/estg/__15.html),
[§ 20 EStG](https://www.gesetze-im-internet.de/estg/__20.html).

Wir rechnen F4 hier für die Personengesellschaft. Beim
Einzelunternehmen wäre ein Darlehen des Inhabers an sich selbst
kein entsprechender Gesellschaftervertrag. Die
Kapitalgesellschaft benötigt wegen der anderen Behandlung von
Gesellschafterdarlehen und Auszahlungen eine eigene Rechnung.

```python
f4_betriebsguthaben = mittel_im_unternehmen_2026
f4_privatguthaben = 0.0

f4_darlehensbetrag = investitionsbetrag
f4_gesellschafterzins = zins_fest_buch

f4_verlauf_zeilen = []

for jahr in steuerplan_kapitalgesellschaft.itertuples(
    index=False
):
    # Beide Ehepartner erhalten zusammen Zinsen
    # auf ihre Darlehen von jeweils 50.000.
    darlehenszinsen_an_gesellschafter = (
        f4_darlehensbetrag
        * f4_gesellschafterzins
    )

    betriebliche_guthabenzinsen = (
        f4_betriebsguthaben
        * guthabenzins_buch
    )

    # Das private Guthaben besteht zu Jahresbeginn
    # aus früher empfangenen Darlehenszinsen.
    private_bankzinsen = (
        f4_privatguthaben
        * guthabenzins_buch
    )

    steuerpflichtige_private_bankzinsen = max(
        0.0,
        private_bankzinsen
        - sparer_pauschbetrag_ehepaar,
    )

    private_kapitalertragsteuer = (
        steuerpflichtige_private_bankzinsen
        * steuersatz_kapitalertraege
    )

    privater_soli = (
        private_kapitalertragsteuer
        * soli_satz_auf_kapitalertragsteuer
    )

    private_steuer_zusammen = (
        private_kapitalertragsteuer
        + privater_soli
    )

    # Gesellschaftlicher Gewinn nach Zinsaufwand
    # plus gleich hohe Sondervergütung der Partner:
    gewinn_nach_darlehenszins = (
        ueberschuss_alt_buch
        + ueberschuss_nominal
        - jahr.afa_buch
        + betriebliche_guthabenzinsen
        - darlehenszinsen_an_gesellschafter
    )

    sonderverguetung_gesellschafter = (
        darlehenszinsen_an_gesellschafter
    )

    gewinn_des_haushalts = (
        gewinn_nach_darlehenszins
        + sonderverguetung_gesellschafter
    )

    zusaetzliche_betriebliche_steuern = (
        modellierte_haushaltssteuer(
            gewinn_des_haushalts
        )["steuern_zusammen"]
        - steuer_altgeschaeft
    )

    tilgung = (
        f4_darlehensbetrag
        if jahr.jahr == 2032
        else 0.0
    )

    f4_betriebsguthaben = (
        f4_betriebsguthaben
        + betriebliche_guthabenzinsen
        + ueberschuss_nominal
        - darlehenszinsen_an_gesellschafter
        - zusaetzliche_betriebliche_steuern
        - tilgung
    )

    f4_privatguthaben = (
        f4_privatguthaben
        + private_bankzinsen
        - private_steuer_zusammen
        + darlehenszinsen_an_gesellschafter
        + tilgung
    )

    f4_verlauf_zeilen.append({
        "jahr": jahr.jahr,
        "betriebliche_guthabenzinsen": (
            betriebliche_guthabenzinsen
        ),
        "darlehenszinsen_an_gesellschafter": (
            darlehenszinsen_an_gesellschafter
        ),
        "sonderverguetung": (
            sonderverguetung_gesellschafter
        ),
        "gewinn_des_haushalts": (
            gewinn_des_haushalts
        ),
        "zusaetzliche_betriebliche_steuern": (
            zusaetzliche_betriebliche_steuern
        ),
        "private_bankzinsen": private_bankzinsen,
        "private_steuer_zusammen": (
            private_steuer_zusammen
        ),
        "tilgung": tilgung,
        "betriebsguthaben_ende": (
            f4_betriebsguthaben
        ),
        "privatguthaben_ende": (
            f4_privatguthaben
        ),
        "gemeinsames_endvermoegen": (
            f4_betriebsguthaben
            + f4_privatguthaben
        ),
    })

f4_verlauf = pd.DataFrame(f4_verlauf_zeilen)

f4_verlauf.round(2)
```

```python
f4_endbetrag = (
    f4_verlauf.loc[
        f4_verlauf["jahr"].eq(2032),
        "gemeinsames_endvermoegen",
    ].iloc[0]
)

ja_haushalt_mit_f4 = pd.concat(
    [
        ja_haushalt_vergleich[
            [
                "finanzierung",
                "endbetrag_2032",
            ]
        ],
        pd.DataFrame([{
            "finanzierung": (
                "F4: Gesellschafterdarlehen "
                "(Personengesellschaft)"
            ),
            "endbetrag_2032": f4_endbetrag,
        }]),
    ],
    ignore_index=True,
)

ja_haushalt_mit_f4[
    "kaufkraft_2026_basis"
] = (
    ja_haushalt_mit_f4[
        "endbetrag_2032"
    ]
    * 100
    / preisindex_basis_2032
)

print(ja_haushalt_mit_f4.round(2))
```

## 24. Wie empfindlich ist F4 gegenüber dem Gesellschafterzins?

> **Frage des Beraters:** „Ändert sich unsere Entscheidung,
> wenn die Gesellschafter für ihr Darlehen einen anderen
> Vertragszins vereinbaren? Und reichen die betrieblichen
> Mittel während der Laufzeit für die Zahlungen aus?“

Wir berechnen F4 erneut mit festen Gesellschafterzinsen von
8 %, 10 %, 12 % und 14 %. Dies sind **hypothetische
Sensitivitätswerte**; wir leiten daraus weder ein
marktübliches Angebot noch die steuerliche Anerkennung
eines konkreten Vertrags ab.

Für jeden Zinsfall bleiben Investition, Projektüberschüsse,
Abschreibung, betrieblicher und privater Guthabenzins,
Steuerannahmen sowie Rückzahlung Ende 2032 gleich.
Die Zinsen des Gesellschafterdarlehens werden bei der
Personengesellschaft weiterhin als Sondervergütungen
der Mitunternehmer behandelt.

Wir prüfen zwei Ergebnisse:

1. **Gemeinsames Endvermögen:** Welchen Betrag besitzen
   Unternehmen und Haushalt zusammen Ende 2032?
2. **Betriebliche Liquidität:** Wie niedrig ist das
   betriebliche Guthaben jeweils am Jahresende? Dieses
   vereinfachte Maß zeigt keine unterjährigen Engpässe.

Ein anderer Gesellschafterzins verteilt zunächst Geld
zwischen Unternehmen und Eigentümern um. Sein Effekt
auf das gemeinsame Endvermögen entsteht hier vor allem
dadurch, dass die verteilten Beträge anschließend
an unterschiedlichen Orten angelegt werden.

```python
def f4_bei_gesellschafterzins(
    gesellschafterzins
):
    """F4 für einen hypothetischen festen Vertragszins."""
    betriebsguthaben = (
        mittel_im_unternehmen_2026
    )
    privatguthaben = 0.0

    verlauf = []

    for jahr in steuerplan_kapitalgesellschaft.itertuples(
        index=False
    ):
        gesellschafterzinsen = (
            investitionsbetrag
            * gesellschafterzins
        )

        betriebliche_zinsen = (
            betriebsguthaben
            * guthabenzins_buch
        )

        private_bankzinsen = (
            privatguthaben
            * guthabenzins_buch
        )

        steuerpflichtige_bankzinsen = max(
            0.0,
            private_bankzinsen
            - sparer_pauschbetrag_ehepaar,
        )

        private_steuer = (
            steuerpflichtige_bankzinsen
            * steuersatz_kapitalertraege
            * (
                1
                + soli_satz_auf_kapitalertragsteuer
            )
        )

        # Zinsaufwand der Gesellschaft und gleich
        # hohe Sondervergütung der Partner werden
        # im zusammengefassten Gewinn berücksichtigt.
        gesellschaftsgewinn = (
            ueberschuss_alt_buch
            + ueberschuss_nominal
            - jahr.afa_buch
            + betriebliche_zinsen
            - gesellschafterzinsen
        )

        gewinn_des_haushalts = (
            gesellschaftsgewinn
            + gesellschafterzinsen
        )

        zusaetzliche_betriebliche_steuer = (
            modellierte_haushaltssteuer(
                gewinn_des_haushalts
            )["steuern_zusammen"]
            - steuer_altgeschaeft
        )

        tilgung = (
            investitionsbetrag
            if jahr.jahr == 2032
            else 0.0
        )

        betriebsguthaben = (
            betriebsguthaben
            + betriebliche_zinsen
            + ueberschuss_nominal
            - gesellschafterzinsen
            - zusaetzliche_betriebliche_steuer
            - tilgung
        )

        privatguthaben = (
            privatguthaben
            + private_bankzinsen
            - private_steuer
            + gesellschafterzinsen
            + tilgung
        )

        verlauf.append({
            "jahr": jahr.jahr,
            "betriebsguthaben": (
                betriebsguthaben
            ),
            "privatguthaben": (
                privatguthaben
            ),
            "gemeinsames_vermoegen": (
                betriebsguthaben
                + privatguthaben
            ),
        })

    return pd.DataFrame(verlauf)


zinsfaelle_f4 = [0.08, 0.10, 0.12, 0.14]

f4_sensitivitaet_zeilen = []

for gesellschafterzins in zinsfaelle_f4:
    verlauf = f4_bei_gesellschafterzins(
        gesellschafterzins
    )

    endvermoegen = (
        verlauf["gemeinsames_vermoegen"]
        .iloc[-1]
    )

    f4_sensitivitaet_zeilen.append({
        "gesellschafterzins_prozent": (
            100 * gesellschafterzins
        ),
        "endvermoegen_2032": (
            endvermoegen
        ),
        "vorteil_gegenueber_f2": (
            endvermoegen - f2_betrag
        ),
        "vorteil_gegenueber_f5": (
            endvermoegen
            - ja_haushalt_vergleich.loc[
                ja_haushalt_vergleich[
                    "finanzierung"
                ].eq("F5"),
                "endbetrag_2032",
            ].iloc[0]
        ),
        "kleinstes_betriebsguthaben_jahresende": (
            verlauf["betriebsguthaben"]
            .min()
        ),
    })

f4_sensitivitaet = pd.DataFrame(
    f4_sensitivitaet_zeilen
)

# Die erneute Rechnung mit 12 % muss das
# bereits berechnete F4-Ergebnis reproduzieren.
assert abs(
    f4_sensitivitaet.loc[
        f4_sensitivitaet[
            "gesellschafterzins_prozent"
        ].eq(12.0),
        "endvermoegen_2032",
    ].iloc[0]
    - f4_endbetrag
) < 0.01

f4_sensitivitaet.round(2)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    f4_sensitivitaet[
        "gesellschafterzins_prozent"
    ],
    f4_sensitivitaet[
        "endvermoegen_2032"
    ],
    marker="o",
    linewidth=2,
    label="F4: Gesellschafterdarlehen",
)

for finanzierung, farbe in [
    ("F1", "#2a6f97"),
    ("F5", "#e07a5f"),
    ("F2", "#6a994e"),
]:
    betrag = (
        ja_haushalt_vergleich.loc[
            ja_haushalt_vergleich[
                "finanzierung"
            ].eq(finanzierung),
            "endbetrag_2032",
        ].iloc[0]
    )

    ax.axhline(
        betrag,
        linestyle="--",
        color=farbe,
        label=finanzierung,
    )

ax.set(
    title=(
        "F4: Einfluss des vereinbarten "
        "Gesellschafterzinses"
    ),
    xlabel="Gesellschafterzins pro Jahr (%)",
    ylabel="Gemeinsames Endvermögen 2032",
)

ax.set_xticks(
    f4_sensitivitaet[
        "gesellschafterzins_prozent"
    ]
)

ax.grid(alpha=0.25)
ax.legend()
plt.tight_layout()
plt.show()
```

## 25. Beratungsübersicht: Entscheidung nach verfügbaren Mitteln

> **Frage des Beraters:** „Stehen Ihnen zusätzlich zu den
> 100.000 im Unternehmen weitere 100.000 privat zur
> Verfügung?“

Die Antwort bestimmt, welche Finanzierungen im Modell
überhaupt möglich sind:

- **Nein:** Wir vergleichen F1 und F5 bei einem gemeinsamen
  Anfangsvermögen von 100.000.
- **Ja:** Wir vergleichen F1, F2, F4 und F5 bei einem
  gemeinsamen Anfangsvermögen von 200.000.

Für die Übersicht legen wir die Personengesellschaft und
den bereits festgelegten gemeinsam veranlagten Haushalt
zugrunde. Die bisherige F1/F5-Rechnung des Haushalts wird
dazu als Rechnung für diese Personengesellschaft gelesen.

Der Rang wird **getrennt für jede Antwort** vergeben.
Die Kaufkraftrechnung verwendet denselben Inflationspfad
für alle Wege. Sie verändert deshalb ihre nominale
Rangfolge nicht.

```python
haushalt_f1_f5 = (
    vergleich_antworten.loc[
        vergleich_antworten["modell"].eq(
            "Einzelunternehmen / "
            "Personengesellschaft"
        ),
        [
            "antwort",
            "finanzierung",
            "endbetrag_2032",
            "kaufkraft_2026_basis",
        ],
    ]
    .copy()
)

zusaetzliche_wege_ja = (
    ja_haushalt_mit_f4.loc[
        ja_haushalt_mit_f4[
            "finanzierung"
        ].str.startswith(("F2", "F4")),
        [
            "finanzierung",
            "endbetrag_2032",
            "kaufkraft_2026_basis",
        ],
    ]
    .copy()
)

zusaetzliche_wege_ja.insert(
    0,
    "antwort",
    "Ja: zusätzliche 100.000 privat",
)

beratungsuebersicht = pd.concat(
    [
        haushalt_f1_f5,
        zusaetzliche_wege_ja,
    ],
    ignore_index=True,
)

beratungsuebersicht["rang_innerhalb_der_antwort"] = (
    beratungsuebersicht
    .groupby("antwort")["endbetrag_2032"]
    .rank(
        ascending=False,
        method="min",
    )
    .astype(int)
)

beratungsuebersicht["abstand_zum_besten_weg"] = (
    beratungsuebersicht["endbetrag_2032"]
    - beratungsuebersicht
        .groupby("antwort")["endbetrag_2032"]
        .transform("max")
)

beratungsuebersicht = (
    beratungsuebersicht
    .sort_values(
        [
            "antwort",
            "rang_innerhalb_der_antwort",
        ]
    )
    .reset_index(drop=True)
)

beratungsuebersicht.round(2)
```

## 26. Kapitalgesellschaft im Ja-Fall: F2 zunächst auf Unternehmensebene

Für die Kapitalgesellschaft untersuchen wir nun die
zusätzliche Einlage F2. Der Haushalt besitzt Ende 2026
zusätzlich 100.000 privat und führt diesen Betrag der
Gesellschaft zu. Die Gesellschaft bezahlt damit die
Investition. Ihre schon vorhandenen 100.000 bleiben
betrieblich angelegt.

Wir berechnen zunächst ausschließlich das Guthaben,
das Ende 2032 **in der Kapitalgesellschaft** verbleibt.
Dafür verwenden wir dieselben Projektüberschüsse,
Guthabenzinsen, Abschreibungen und vereinfachten
Körperschaft- und Gewerbesteuerannahmen wie in den
bisherigen Kapitalgesellschaftsrechnungen.

**Grenze dieses Schritts:** Eine neue Einlage ist nicht
automatisch eine später steuerfrei auszahlbare Summe.
Ebenso darf nicht das gesamte Guthaben pauschal als
Gewinnausschüttung besteuert werden. Für einen Vergleich
aus Eigentümersicht wären insbesondere die Einlage,
das steuerliche Einlagekonto, ausschüttbare Gewinne
und die tatsächliche Art der Auszahlung zu bestimmen.
Die folgende Ausgabe ist deshalb ein Ergebnis **auf
Unternehmensebene**, noch keine Rechtsformempfehlung.

Rechtsgrundlage zur Einlagenrückgewähr:
[§ 27 KStG](https://www.gesetze-im-internet.de/kstg_1977/__27.html).

```python
f2_kg_guthaben = (
    mittel_im_unternehmen_2026
)

f2_kg_zeilen = []

for jahr in steuerplan_kapitalgesellschaft.itertuples(
    index=False
):
    # Die zusätzliche Einlage bezahlt die Investition.
    # Das bereits vorhandene Unternehmensguthaben
    # bleibt als Anfangsguthaben bestehen.
    guthabenzinsen = (
        f2_kg_guthaben
        * guthabenzins_buch
    )

    gewinn_mit_vorhaben = (
        ueberschuss_alt_buch
        + ueberschuss_nominal
        - jahr.afa_buch
        + guthabenzinsen
    )

    zusaetzliche_kst = (
        gewinn_mit_vorhaben
        - ueberschuss_alt_buch
    ) * jahr.kst_satz_prozent / 100

    zusaetzliche_gewerbesteuer = (
        modellierte_gewerbesteuer_kapitalgesellschaft(
            gewinn_mit_vorhaben
        )
        - modellierte_gewerbesteuer_kapitalgesellschaft(
            ueberschuss_alt_buch
        )
    )

    f2_kg_guthaben = (
        f2_kg_guthaben
        + guthabenzinsen
        + ueberschuss_nominal
        - zusaetzliche_kst
        - zusaetzliche_gewerbesteuer
    )

    f2_kg_zeilen.append({
        "jahr": jahr.jahr,
        "guthabenzinsen": (
            guthabenzinsen
        ),
        "zusaetzliche_kst": (
            zusaetzliche_kst
        ),
        "zusaetzliche_gewerbesteuer": (
            zusaetzliche_gewerbesteuer
        ),
        "guthaben_in_gesellschaft": (
            f2_kg_guthaben
        ),
        "neue_einlage_2026": (
            investitionsbetrag
        ),
    })

f2_kg_verlauf = pd.DataFrame(
    f2_kg_zeilen
)

f2_kg_verlauf.round(2)
```

### 26.1 Ergebnis F2: Guthaben in der Kapitalgesellschaft

Bei F2 verbleiben Ende 2032 **325.254,58 Geldeinheiten**
im betrachteten Guthaben der Kapitalgesellschaft. Darin
wirken die schon vorhandenen 100.000 betrieblichen Mittel,
die neue Investition, die jährlichen Projektüberschüsse,
Guthabenzinsen und die modellierten zusätzlichen Steuern
zusammen.

Der Betrag ist keine Nettoauszahlung an den Haushalt.
Die Eigentümer haben für F2 zusätzlich 100.000 eingebracht.
Ob und wie Gesellschaftsmittel später an sie ausgezahlt
werden, ist in dieser Rechnung noch nicht festgelegt.
Deshalb ordnen wir F2 nicht anhand dieses Betrags in
die bisherige Rangfolge der Nettoauszahlungen ein.

## 27. F4 bei der Kapitalgesellschaft: Gesellschafterdarlehen

> **Frage des Beraters:** „Was ändert sich, wenn die
> zusätzlichen privaten 100.000 als Darlehen statt als
> Einlage in die Kapitalgesellschaft fließen?“

Beide Ehepartner halten im Modell jeweils 50 % der
Kapitalgesellschaft und leihen ihr jeweils 50.000.
Die Gesellschaft finanziert damit die Investition;
ihre bereits vorhandenen 100.000 bleiben angelegt.

Wir übernehmen für einen kontrollierten Vergleich
den festen Kreditzins von 12 %, die jährliche
Zinszahlung und die vollständige Tilgung Ende 2032
aus dem bisherigen Referenzfall. Diese 12 % sind
eine Annahme aus dem historischen Buchbeispiel,
kein geprüftes Vertragsangebot.

**Unternehmensebene:** Wir berechnen das Guthaben
der Kapitalgesellschaft nach den bislang modellierten
zusätzlichen Körperschaft- und Gewerbesteuern.

**Eigentümerebene:** Wir zeigen die empfangenen
Darlehenszinsen und die Tilgung als Zahlungen vor
persönlicher Steuer. Eine Nettoauszahlung wird
hieraus noch nicht berechnet. Für Darlehenszinsen
an wesentlich beteiligte Anteilseigner kann der
pauschale Steuertarif ausgeschlossen sein; die
bisherige private Zinsfunktion wäre hierfür keine
geeignete Abkürzung.

Rechtsgrundlage:
[§ 32d Abs. 2 EStG](https://www.gesetze-im-internet.de/estg/__32d.html).

```python
f4_kg_guthaben = (
    mittel_im_unternehmen_2026
)

f4_kg_darlehen = (
    investitionsbetrag
)

f4_kg_zeilen = []

for jahr in steuerplan_kapitalgesellschaft.itertuples(
    index=False
):
    guthabenzinsen = (
        f4_kg_guthaben
        * guthabenzins_buch
    )

    zinsen_an_gesellschafter = (
        f4_kg_darlehen
        * zins_fest_buch
    )

    gewinn_mit_vorhaben = (
        ueberschuss_alt_buch
        + ueberschuss_nominal
        - jahr.afa_buch
        + guthabenzinsen
        - zinsen_an_gesellschafter
    )

    zusaetzliche_kst = (
        gewinn_mit_vorhaben
        - ueberschuss_alt_buch
    ) * jahr.kst_satz_prozent / 100

    zusaetzliche_gewerbesteuer = (
        modellierte_gewerbesteuer_kapitalgesellschaft(
            gewinn_mit_vorhaben
        )
        - modellierte_gewerbesteuer_kapitalgesellschaft(
            ueberschuss_alt_buch
        )
    )

    tilgung_an_gesellschafter = (
        f4_kg_darlehen
        if jahr.jahr == 2032
        else 0.0
    )

    f4_kg_guthaben = (
        f4_kg_guthaben
        + guthabenzinsen
        + ueberschuss_nominal
        - zinsen_an_gesellschafter
        - zusaetzliche_kst
        - zusaetzliche_gewerbesteuer
        - tilgung_an_gesellschafter
    )

    f4_kg_zeilen.append({
        "jahr": jahr.jahr,
        "guthabenzinsen": guthabenzinsen,
        "zinsen_an_gesellschafter_vor_steuer": (
            zinsen_an_gesellschafter
        ),
        "zusaetzliche_kst": (
            zusaetzliche_kst
        ),
        "zusaetzliche_gewerbesteuer": (
            zusaetzliche_gewerbesteuer
        ),
        "tilgung_an_gesellschafter": (
            tilgung_an_gesellschafter
        ),
        "guthaben_in_gesellschaft": (
            f4_kg_guthaben
        ),
    })

f4_kg_verlauf = pd.DataFrame(
    f4_kg_zeilen
)

f4_kg_verlauf.round(2)
```
